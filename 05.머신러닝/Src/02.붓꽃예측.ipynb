## 교차검증
* 데이터를 여러 부분으로 나누어 학습과 검증을 반복하면서 특정 데이터에 대한 평가 편향을 줄임
* 모델의 일반화 성능과 안정적 평가를 위해 사용

```python
dt_clf = DecisionTreeClassifier()
iris_data = load_iris()

X_train, X_test, y_train, y_test = train_test_split(
    iris_data.data, iris_data.target, test_size=0.3, random_state=12  # 200, 42?
)

dt_clf.fit(X_train, y_train)
pred = dt_clf.predict(X_test)
print("예측 정확도: {0:.4f}".format(accuracy_score(y_test, pred)))
```

* K 폴드

```python
from sklearn.model_selection import KFold

iris = load_iris()
features = iris.data
label = iris.target
dt_clf = DecisionTreeClassifier(random_state=12)

kfold = KFold(n_splits=5)   # shuffle=False(Default), shuffle=True일 경우 데이터 무작위로 섞음(y값은 상관 안함)
print("붓꽃 데이터 세트 크기:", features.shape[0])
```

```python
import numpy as np

n_iter = 0
cv_accuracy = []  # split별로 평균 저장 

for train_index, test_index in kfold.split(features):
    X_train, X_test = features[train_index], features[test_index]
    y_train, y_test = label[train_index], label[test_index]

    # 학습 및 예측
    dt_clf.fit(X_train, y_train)
    pred = dt_clf.predict(X_test)
    n_iter += 1

    # 반복 시 마다 정확도 측정
    accuracy = np.round(accuracy_score(y_test, pred), 4)
    train_size = X_train.shape[0]
    test_size = X_test.shape[0]

    print(
        "\n#{0} 교차 검증 정확도 :{1}, 학습 데이터 크기: {2}, 검증 데이터 크기: \
            {3}".format(n_iter, accuracy, train_size, test_size)
    )

    print("#{0} 검증 세트 인덱스:{1}".format(n_iter, test_index))
    cv_accuracy.append(accuracy)
    print()
    
print("\n## 평균 검증 정확도:", np.mean(cv_accuracy))
```

* Stratified K 폴드

```python
import pandas as pd

iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df["label"] = iris.target

print(iris_df["label"].value_counts())
print()
iris_df.head()
```

```python
kfold = KFold(n_splits=3)
n_iter = 0

for train_index, test_index in kfold.split(iris_df):
    n_iter += 1
    label_train = iris_df["label"].iloc[train_index]
    label_test = iris_df["label"].iloc[test_index]
    print("## 교차 검증: {0}".format(n_iter))
    print("학습 레이블 데이터 분포:\n", label_train.value_counts())
    print("검증 레이블 데이터 분포:\n", label_test.value_counts())

# 문제점 : # y값이 골고루 섞이지 않음
```

```python
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits=3)
n_iter = 0

for train_index, test_index in skf.split(iris_df, iris_df["label"]):
    n_iter += 1
    label_train = iris_df["label"].iloc[train_index]
    label_test = iris_df["label"].iloc[test_index]
    print("## 교차 검증: {0}".format(n_iter))
    print("학습 레이블 데이터 분포:\n", label_train.value_counts())
    print("검증 레이블 데이터 분포:\n", label_test.value_counts())
```

```python
dt_clf = DecisionTreeClassifier(random_state=156)

skfold = StratifiedKFold(n_splits=3)
n_iter = 0
cv_accuracy = []

for train_index, test_index in skf.split(iris_df, iris_df["label"]):
    n_iter += 1
    X_train, X_test = features[train_index], features[test_index]
    y_train, y_test = label[train_index], label[test_index]

    dt_clf.fit(X_train, y_train)
    pred = dt_clf.predict(X_test)

    n_iter += 1
    accuracy = np.round(accuracy_score(y_test, pred), 4)
    cv_accuracy.append(accuracy)

print("\n## 교차 검증별 정확도:", np.round(cv_accuracy, 4))
print("## 평균 검증 정확도:", np.mean(cv_accuracy))
```

* cross_val_score( )
    * 폴드 세트 설정, for 루프에서 반복적으로 학습/테스트 데이터의 인덱스를 추출한 뒤 반복적으로 학습과 예측을 수행하고 예측 성능들을 반환해서 평균을 냄
    * 위 일련의 과정을 한꺼번에 수행해주는 API 
    * estimator가 분류면 y값을 기준으로 섞어도 되므로 stratified k-fold 수행
    * estimator가 회귀면 y값을 기준으로 섞으면 안되며르 k-fold 수행

```python
from sklearn.model_selection import cross_val_score

iris_data = load_iris()
dt_clf = DecisionTreeClassifier(random_state=156)

data = iris_data.data
label = iris_data.target

scores = cross_val_score(dt_clf, data, label, scoring="accuracy", cv=3)
print("교차 검증별 정확도:", np.round(scores, 4))
print("평균 검증 정확도:", np.round(np.mean(scores), 4))
```

* GridSearchCV(교차검증 + 하이퍼 파라미터 튜닝)

```python
from sklearn.model_selection import GridSearchCV

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=121
)

dtree = DecisionTreeClassifier()
```

```python
# max_depth : 더이상 훈련을 진행하지 않을 트리의 깊이 
# min_samples_split : 훈련시 분할할 수 있는 노드의 최소 샘플수 
parameters = {"max_depth": [1, 2, 3], "min_samples_split": [2, 3]}

# refit=True : 찾은 최적의 모델의 파라메터값에 실제 사용할 최종모델을 만들기 위해 120개의 데이터를 전부 다시 돌리는 과정(best_estimator_ 추출됨)
grid_dtree = GridSearchCV(dtree, param_grid=parameters, cv=3, refit=True)
grid_dtree.fit(X_train, y_train)

scores_df = pd.DataFrame(grid_dtree.cv_results_)
scores_df[
    [
        "params",
        "mean_test_score",
        "rank_test_score",
        "split0_test_score",
        "split1_test_score",
        "split2_test_score",
    ]
]
```

```python
print("GridSearchCV 최적 파라미터:", grid_dtree.best_params_)
print("GridSearchCV 최고 정확도: {0:.4f}".format(grid_dtree.best_score_))
```

```python
estimator = grid_dtree.best_estimator_
pred = estimator.predict(X_test)
print("테스트 데이터 세트 정확도: {0:.4f}".format(accuracy_score(y_test, pred)))
```
