## 01. 수익률 등급 분류

매수가와 현재가가 주어질 때 수익률을 계산하고 등급을 A/B/C/D로 분류하시오. (A: 15% 이상, B: 5% 이상, C: 0% 이상, D: 그 외)

```python
buy_price = 80000      # 매수가 (주식을 산 가격)
current_price = 92000  # 현재가 (지금 주식 가격)

# 수익률 계산 후 조건문으로 등급을 결정하세요.
# 수익률(%) = (현재가 - 매수가) / 매수가 * 100
profit_rate = (current_price - buy_price) / buy_price * 100

# 위에서부터 차례로 검사하므로 높은 기준부터 비교해야 함
if profit_rate >= 15:    # 15% 이상
    grade = 'A'
elif profit_rate >= 5:   # 5% 이상 15% 미만
    grade = 'B'
elif profit_rate >= 0:   # 0% 이상 5% 미만
    grade = 'C'
else:                    # 0% 미만 (손실)
    grade = 'D'

# :.2f -> 소수점 둘째 자리까지 출력
print(f'수익률: {profit_rate:.2f}%')
print(f'등급: {grade}')
```

## 02. 목표 수익률 매도가

매수가와 목표 수익률이 주어질 때 목표 매도가를 계산하시오.

```python
buy_price = 75000   # 매수가
target_rate = 0.12  # 목표 수익률 (12%)

# 목표 매도가를 계산하세요.
# 목표 매도가 = 매수가 * (1 + 목표 수익률)
# 예) 75,000원에 12% 수익 -> 75,000 * 1.12
target_price = buy_price * (1 + target_rate)

# :,.0f -> 천 단위 콤마 + 소수점 없이 출력
print(f'목표 매도가: {target_price:,.0f}원')
```

## 03. 손익분기점 계산

매수가, 수수료율, 세율이 주어질 때 손익분기점 매도가를 계산하시오. 매도금액에 수수료와 세금이 발생한다고 가정합니다.

- 매도가 × (1 - 수수료율 - 세율) = 매수가
- 매도가 = 매수가 / (1 - 수수료율 - 세율)

```python
buy_price = 100000  # 매수가
fee_rate = 0.001    # 매도 수수료율 (0.1%)
tax_rate = 0.002    # 매도 세율 (0.2%)

# 손익분기점 매도가를 계산하세요.
# 매도 시 수수료와 세금을 떼고 실제로 받는 금액 = 매도가 * (1 - 수수료율 - 세율)
# 이 금액이 매수가와 같아지는 매도가가 손익분기점
# -> 매도가 = 매수가 / (1 - 수수료율 - 세율)
break_even_price = buy_price / (1 - fee_rate - tax_rate)

# 이 가격보다 비싸게 팔아야 실제로 이익이 남음
print(f'손익분기점 매도가: {break_even_price:,.2f}원')
```

## 04. 월 복리 계산

원금이 연 수익률로 월 복리 운용될 때 N개월 후 금액을 반복문으로 계산하시오.

```python
principal = 10000000  # 원금 (1,000만 원)
annual_rate = 0.06    # 연 수익률 (6%)
months = 24           # 운용 기간 (24개월)

# 월 복리로 계산하세요.
monthly_rate = annual_rate / 12  # 연 수익률을 12로 나눠 월 수익률로 변환 (0.5%)
amount = principal               # 현재 금액을 원금으로 시작

# 1개월부터 24개월까지 반복하며 매달 이자를 원금에 더함 (복리)
for month in range(1, months + 1):
    amount = amount * (1 + monthly_rate)  # 이번 달 금액 = 지난달 금액 * (1 + 월 수익률)

print(f'{months}개월 후 금액: {amount:,.0f}원')
print(f'수익금: {amount - principal:,.0f}원')  # 최종 금액 - 원금 = 수익금
```
