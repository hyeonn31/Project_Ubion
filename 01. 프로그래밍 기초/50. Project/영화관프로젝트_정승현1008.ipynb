```python
## 1. 예매할 영화 선택하기
movies = ["Dark Knight", "Harry Porter", "Parasite", "Matrix", "La La Land"]
print(f'{"영화목록":=^30}')
for i in movies:
    print(i)
print("=" * 33)

choice = input("예매하실 영화를 선택해주세요: ")

while choice not in movies:
    print("예약할 수 없는 영화입니다.")
    choice = input("예매하실 영화를 선택해주세요: ")
    
print(f"{choice}영화를 선택하셨습니다.")
```

```python
## 2. 관람 인원 수 입력받기
while True:
    people_text = input("인원 수를 입력하세요: ")

    if not people_text.isdigit():            # 숫자가 아니면 (abc, -1, 2.5 등)
        print("관람 인원 수는 숫자로만 입력하세요.")
    elif int(people_text) <= 0:              # 0이면
        print("관람 인원 수는 1명 이상이어야 합니다.")
    else:
        people = int(people_text)
        print(f"관람할 인원수는 {people}명 입니다.")
        break                                # 올바른 값이면 반복 종료
```

```python
## 3. 할인권 여부 확인하기
discount_dic = {"ORANGE": 2000, "VALENTINE": 3000, "CHRISTMAS": 4000, "INDEPENDENCE": 5000}
discount = 0   # 할인 금액 (기본은 0원)

use_coupon = input("할인권을 사용하시려면 y, 금액 확인으로 넘어가시려면 n 을 입력하세요")

if use_coupon == "y":
    coupon = input("할인권 코드를 입력하세요").strip().upper()

    if coupon in discount_dic:             # 할인권 코드가 존재하면
        discount = discount_dic[coupon]    # 그 코드의 가격을 새 변수에 저장
        print(f"{coupon} 할인권 적용: {discount}원 할인")
    else:
        print("존재하지 않는 할인권입니다.")

print("할인 금액:", discount)
```

```python
## 4. 예매 티켓 가격 출력하기

original_price = people * 12000
price_sum = original_price - discount

print(f"인원: {people}명")
print(f"정가: {original_price:,}원")
print(f"할인: -{discount:,}원")
print(f"결제 금액: {price_sum:,}원")
```
