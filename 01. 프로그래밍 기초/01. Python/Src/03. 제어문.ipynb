### 1. 조건문(if)

```python
# 가상의 계좌에 주문 가능 현금이 있는가?

has_cash = True

if has_cash:
    print('매수 주문을 검토합니다.')
    print('주문 가능 금액을 확인합니다. ')
else:
    print("매수 주문을 내지 않습니다.")
    print("잔액을 먼저 확인합니다.")
    
```

### 비교연산자 : >,<,<=,>=,==,!=

```python
close_price = 51000
reference_price = 50000

print(close_price < reference_price)
print(close_price > reference_price)
print(close_price == reference_price)
print(close_price != reference_price)
print(close_price <= reference_price)
print(close_price >= reference_price)
```

```python
# 주문을 하려고 하는데 주문 금액 조건이 충족되면 '가상 주문의 금액 조건 충족' 출력
# 미충족인 경우 '가상 주문의 금액 조건 미충족' 출력

cash= 20000
order_cost = 30000

if cash >= order_cost :
    print("가상 주문의 금액 조건 충족")
else :
    print("미충족")
```

### 2. 반복문(for)

```python
asset_codes = ['ETF','BOND','CASH']
for asset_code in asset_codes:
    print("관찰자산: ",asset_code, end=' / ')
print()
```

```python
# ??일차 수익률(양수)
daily_returns = [0.03, -0.02, 0.01, -0.01, 0.04]

for day, r in enumerate(daily_returns, start=1):
    if r > 0:
        print(f"{day}일차 수익률: {r:.2%}")
```

```python
for day, r in enumerate(daily_returns):
    if r > 0:
        print("일차",day+1, "수익률:",r)
```

```python
daily_returns = [0.03,-0.02, 0.01, -0.01,0.04]
for i in range(len(daily_returns)):
    if daily_returns[i] <=0:
        continue
    print("%d일차 수익률 : %.3f" % (i+1,daily_returns[i]))
```

### 3. 리스트 내포(list comprehension)

```python
base_deposits = [1000, 2000, 3000, 4000]
base_deposits.append(5000)
print(base_deposits)
```

```python
# 기존 데이터의 3배수를 해서 tripled 리스트에 추가하기
base_deposits = [1000, 2000, 3000, 4000]

tripled = []

for d in base_deposits:
    tripled.append(d * 3)

print(tripled)   # [3000, 6000, 9000, 12000]


```

```python
base_deposits = [1000, 2000, 3000, 4000]
tripled = [amount * 3 for amount in base_deposits]
print('리스트 내포:', tripled)
```

```python
base_deposits = [1000, 2000, 3000, 4000]
tripled = [amount * 3 for amount in base_deposits if(amount // 1000)%2 ==0]
print('리스트 내포:', tripled)
```

```python
# 자산종류 : SPY,QQQ,AAPL,TSLA,BOND,GOLD,CASH
# tech skills : while문

watch_list = []

while True:
    print("1. Add  2. Del  3. List  4. Quit")
    menu = input("메뉴 번호를 입력하세요: ")

    if menu == "1":
        asset = input("추가할 자산 코드를 입력하세요: ")
        watch_list.append(asset)
        print("추가:", asset)

    elif menu == "2":
        asset = input("삭제할 자산 코드를 입력하세요: ")
        if asset in watch_list:
            watch_list.remove(asset)
            print("삭제:", asset)
        else:
            print("목록에 없는 자산입니다.")

    elif menu == "3":
        print("현재 관심 자산:", watch_list)

    elif menu == "4":
        print("메뉴 종료")
        break

    else:
        print("잘못된 번호입니다.")

    print()   # 메뉴 사이 빈 줄
```

```python
# 자산종류 : SPY,QQQ,AAPL,TSLA,BOND,GOLD,CASH

w_list=[]

while True :

    print("1. Add 2. Del 3. List 4. Quit")
    input_list = input("메뉴 번호를 입력하세요:")
    
    if input_list =="1": 
        put_num = input("추가할 자산 코드를 입력하세요:")
        w_list.append(put_num)
        print("추가:",put_num)
    
    elif input_list =="2":
        put_num = input("삭제할 자산 코드를 입력하세요:")
        if put_num in w_list:
            w_list.remove(put_num)
            print("삭제 완료: ",put_num)
        else:
            print("해당 자산이 없어요")
            
    elif input_list =="3":
        print("현재 관심자산: ",w_list)
    
    elif input_list == "4":
        print("종료합니다.")
        break
    
    else:
        print("1~4중에 선택하세요")
    
```
