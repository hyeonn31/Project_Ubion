## 함수를 호출하는 방법
- 위치인수(value), 키워드 인수(변수)
- 인수 = 매개변수 = 파라미터 

```python
def print_order_cost(unit_price, quantity):
    total = unit_price * quantity

    return total

print("위치 인수:", print_order_cost(50000,3))
print("키워드 인수:", print_order_cost(unit_price=50000, quantity=3))
print("키워드 순서 변경:",print_order_cost(quantity=3,unit_price=50000) )
```

## 임의 개수의 위치 인수: * args
* *args로 받은 인자값들은 튜플로 구성함
* 매개변수 값들의 개수는 고정이 아님

```python
def sum_deposits(*args):
    print("자료형:", type(args),"| 전달된 값:",args)
    total = 0
    
    for amount in args:
        total += amount
    return total
    
print("3건 합계:", sum_deposits(10000,20000,30000))
print("3건 합계:", sum_deposits(10000,20000,30000,40000))
```

# 계산기에 choice를 먼저 주고, 뒤의 여러 수를 더하거나 곱
    # 함수명 : calculate_many
    # 파라미터 : choice, *args
    # choice : add, mul
    # ex : calculate_many('add',10000,20000,30000)


```python
def calculate_many(choice, *args):
    if choice == 'add':
        result = 0
        for n in args:
            result = result + n
        return result

    elif choice == 'mul':
        result = 1
        for n in args:
            result = result * n
        return result

    else:
        return "add 또는 mul만 입력하세요"

print(calculate_many('add', 10000, 20000, 30000))   # 60000
print(calculate_many('mul', 2, 3, 4))               # 24
```

```python
def calculate_many(choice, *args):
    if choice == 'add':
        result = 0
        for n in args:
            result = result + n
        return result

    elif choice == 'mul':
        result = 1
        for n in args:
            result = result * n
        return result

    else:
        return "add 또는 mul만 입력하세요"


# 1. 계산 방법 입력받기
print("숫자를 선택하세요: 1. add, 2. mul")
menu = input("=> ")

if menu == "1":
    choice = 'add'
elif menu == "2":
    choice = 'mul'
else:
    choice = ''

# 2. 숫자 여러 개 입력받기 (띄어쓰기로 구분)
nums_text = input("숫자들을 띄어쓰기로 입력하세요: ")   # 예: 10000 20000 30000
nums = [int(n) for n in nums_text.split()]
# 3. 함수 호출
print("결과:", calculate_many(choice, *nums))
```

## lambda함수
* 간단한 함수를 짧게 만드는 문법   


```python
# lambda버전
calculate_amount = lambda unit_price, quantity: unit_price * quantity
print('lambda결과:', calculate_amount(50000,3))

# def버전
def calculate_amount2(unit_price, quantity):
    result = unit_price * quantity
    
    return result

print('lambda결과2:', calculate_amount2(50000,3))
```


## 연습문제

```python
# 연습문제 1
# 가격과 수량을 받아 가상 주문 금액을 반환하는 order_total()함수를 작성
# return 없이 바로 출력(함수 내에서)
# return 해서 call되는 곳에서 변수로 받아 출력

order_total = lambda price, quantity: price * quantity
print("총 가격은", order_total(50000,3),"입니다. ")

def order_total2(price, quantity):
    a = price * quantity
    return a

b= order_total2(50000,2)
print(f"총 가격은 {b} 입니다. ")
```

```python
# 연습문제 2 : *args
# 입금액을 몇 개든 받아 합계를 반환하는 total_deposits(*amounts)를 작성하고 세 금액의 합 호출

def total_deposits(*amounts):
    total = 0
    for amount in amounts:
        total += amount
    return total

sum_price = total_deposits(10000, 20000, 30000)
print("세 금액의 합: ", sum_price)
```

```python
# 연습문제 3: **kwargs
#asset, quantity 등을 키워드 인수로 받아 딕셔너리를 반환하는 make_order(**kwargs)를 작성

def make_order(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
    return kwargs

order = make_order(asset='ETF', quantity=3)
```

```python
# 연습문제 4 : 함수 매개변수 기본값
# currency = 'KRW' 기본값을 가지는 format_amount(amount, currency='KRW')를 작성해 기본, 지정 값을 시험

def format_amount(amount, currency='KRW'):
    return str(amount) + " " + currency

print(format_amount(50000))
print(format_amount(50, currency='USD'))



```
