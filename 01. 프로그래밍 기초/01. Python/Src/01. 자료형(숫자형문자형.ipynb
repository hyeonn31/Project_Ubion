### 정렬과 공백

```python
#소수점 표현
"%.3f" %3.42134234
```

```python
"%5.3f" %3.42134234
```

### format 함수

```python
"I eat {0} apples".format(3)
```

```python
"i eat {0} apples. so i was sick for {1} days.".format(3, 5)
```

```python
"i ate {number} apples. so i was sick for {day} days.".format(number=10, day="three")
```

```python
"{0:<10}".format("hi") #왼쪽 정렬
```

```python
#공백채우기
"{0:=>10}".format("hi") #오른쪽 정렬
```

```python
#공백채우기
"{0:=^10}".format("hi") #가운데 정렬
```

```python
# 소수점 표현
y = 3.42134234
"{0:0.4f}".format(y)
```

```python
# { 또는 } 문자표현
"{{ and }}".format()
```

### f문자열 포메팅 : 파이썬 3.6 이상에서 사용 가능

```python
number = 3
day = 5
f"i ate {number} apples. so i was sick for {day} days."
```

```python
name = "홍길동"
age = 30
f"나의 이름은 {name}입니다. 나이는 {age}입니다."
```

```python
f"나의 이름은 {name}입니다. 나이는 {age}입니다. 내년이면 {age+1}살이 됩니다."
```

```python
#정렬
print(f'{"hi":<10}') #왼쪽 정렬
print(f'{"hi":>10}') #오른쪽 정렬
print(f'{"hi":^10}') #가운데 정렬
```

```python
#공백채우기
print(f'{"hi":=^10}') #등호로 채우기
print(f'{"hi":!<10}') #느낌표로 채우기
```

```python
#소수점 표현
y = 3.42134234
f'{y:0.4f}',f'{y:10.4f}'
```

```python
f'{{ and }}'
```

### 문자열 관련 함수들

```python
# 문자 개수 세기 - count
a = "hobby"
a.count("b")
```

```python
# 위치 알려 주기 1 - find
# 결과값이 -1이면 찾지 못했다는 뜻이다. 
a = "Python is the best choice"
a.find("b"),a.find("k")
```

```python
# 위치 알려 주기 2 - index
a = "Life is too short"
a.index("t") 
a.index("k") #찾지 못하면 오류 발생
```

```python
#문자열 삽입 - join
",".join('abcd') # 'a,b,c,d'
```

```python
# 리스트에 join을 사용하면 리스트의 각 요소 사이에 문자열을 삽입할 수 있다.
"#".join(['a', 'b', 'c', 'd']) # 'a#b#c#d'
```

```python
#대소문자 바꾸기
a = "hi"
b = "Hi"
print(a.upper()) #대문자로 변환
print(b.lower()) #소문자로 변환
```

```python
# 공백 지우기
a = " hi "
print(a.lstrip()) #왼쪽 공백 지우기
print(a.rstrip()) #오른쪽 공백 지우기
print(a.strip()) #양쪽 공백 지우기
```
