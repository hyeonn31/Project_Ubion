## 문자열 포메팅

```python
#방법 1 : %
#방법 2 : format()
#방법 3 : f-sㅠtring (파이썬 3.6버전 이상)
```

```python
#숫자대입
a= "i eat %d apples." %3
a
```

```python
#숫자대입
a = "i eat %f apples." %3.4
a
```

```python
#문자대입
a = "i eat %s apples." %"five"
a
```

```python
#숫자 변수로 대입
num = 3
a = "i eat %d apples." %num
a
```
