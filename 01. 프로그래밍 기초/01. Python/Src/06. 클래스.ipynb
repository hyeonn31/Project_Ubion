```python
class Calculator:
    
    def __init__(self):   # 객체를 만드는 순간 자동으로 한 번 실행되는 (생성자)함수
        self.result = 0
        print("test")
    
    def add(self, num):   # 이 계산기 자신을 가리킴
        self.result += num
        return self.result
        
    
    def sub(self, num):
        self.result -= num
        return self.result
                
    
    
```

```python
cal1 = Calculator()
cal2 = Calculator()
```

```python
print(cal1.add(3))


print(cal2.sub(3))
```


#### 가상 계좌 만들기

```python
class PracticeAccount:
    def __init__(self):
        self.balance = 0
        
    def deposit(self, amount):
        self.balance += amount
        return self.balance
    
    def withdraw(self,amount):
        self.balance -= amount
        return self.balance
```

```python
account_a = PracticeAccount()
account_b = PracticeAccount()
```

```python
print("A 입금", account_a.deposit(30000))
print("A 추가 입금", account_a.deposit(40000))
print("A 출금", account_a.withdraw(40000))
print("B 추가 입금", account_b.deposit(40000))
print("B 추가 입금", account_b.deposit(40000))
```

```python
print("서로 독립된 잔액:", account_a.balance, account_b.balance)
```

## setdata()로 속성을 설정하는 계산기


```python
class DepositPair:
    def setdata(self, first, second):
        self.first = first
        self.second = second

    def add(self):
        return self.first + self.second
```

```python
# 30000, 40000 값을 더하기

pair = DepositPair()           # 객체 만들기
pair.setdata(30000, 40000)     # 값 넣기
print(pair.add())              # 70000
```

```python
class DepositPair1:
    def __init__(self):
        self.first = 0
        self.second = 0
        
    def setdata(self, first, second):
        self.first = first
        self.second = second

    def add(self):
        return self.first + self.second
```

```python
# 30000, 40000 값을 더하기

pair = DepositPair1()           # 객체 만들기    # 값 넣기
print(pair.add()) 
pair.setdata(30000,40000)
print(pair.add())
pair.setdata(3000,4000)
print(pair.add())
```

```python
class FinanceCalculator:
    def __init__(self, first, second):
        self.first = first
        self.second = second
        
    def setdata(self, first, second):
            self.first = first
            self.second = second
    
    def add(self):
        return self.first + self.second
    
    def sub(self):
            return self.first - self.second
    
    def mul(self):
            return self.first * self.second
        
    def div(self):
            return self.first / self.second
```

```python
Calculator = FinanceCalculator(30000,40000)
print("두 금액 합계 : " ,Calculator.add())
print("두 금액 차 : " ,Calculator.sub())
```

```python
Calculator.setdata(50000,3)
print("가상 단가 x 수량: " ,Calculator.mul())
print("가상 단가 / 건수: " ,Calculator.div())
```

```python
calculator_a = FinanceCalculator(30000,40000)
calculator_b = FinanceCalculator(50000,3)

print("서로 다른 객체인가?:",calculator_a is not calculator_b)
print("id가 다른가?:",id(calculator_a) != id(calculator_b))
print("각 객체의 피연산자:",(calculator_a.first,calculator_a.second),(calculator_b.first, calculator_b.second))
```

## 클래스 상속(inheritance)

```python
class GrowthCalculator(FinanceCalculator):
    def growth_factor(self): # 성장계수
        return self.first ** self.second

extended = GrowthCalculator(10000,20000)
print("상속 받은 add(가상입금액 합계):",extended.add())

extended.setdata(1.05,2)
print("추가한 growth_factor:",extended.growth_factor())
```


## 메서드 오버라이딩(overriding)

```python
ordinary = FinanceCalculator(50000,0)
ordinary.div()
```

```python
class SafeFinanceCalculator(FinanceCalculator):
    def div(self):
        if self.second ==0:
            return None
        return self.first / self.second
    
safe = SafeFinanceCalculator(50000,0)
print("자식 메서드가 반환한 계산 불가 값:",safe.div())
safe.setdata(50000,2)
print("분모가 0이 아닐 때:",safe.div())
```

## 클래스 변수와 인스턴스 속성
* 클래스 변수는 클래스에 저장되며 인스턴스가 공통으로 조회
* 클래스에서 바꾸면 별도 값이 없는 인스턴스의 조회 결과도 달라짐


```python
class Report:
    report_title = "가상 금융 보고서" # 클래스 변수
```

```python
a = Report()
b = Report()
```

```python
print("클래스 변수: ",Report.report_title)
```

```python
print("두 인스턴스의 조회:",a.report_title,b.report_title)
print("같은 클래스 속성을 참조하는가?:",id(a.report_title)==id(b.report_title))
```

```python
a.report_title = "a만의 가상 보고서"
print("a에서만 속성을 덮어쓴 후:", a.report_title , b.report_title)
print("서로 다른 값을 참조하는가?:",id(a.report_title)!=id(b.report_title))
```

```python
print("a에만 인스턴스 속성이 있는가?", 'report_title' in a.__dict__,
      'report_title' in b.__dict__)
```
