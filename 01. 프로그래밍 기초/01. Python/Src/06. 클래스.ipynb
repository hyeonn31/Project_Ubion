```python
class Calculator:
    
    def __init__(self):   # 객체를 만드는 순간 자동으로 한 번 실행되는 (생성자)함수
        self.result = 0
        print("test")
    
    def add(self, num):   # 이 계산기 자신을 가리킴
        self.result += num
        return self.result
        
    
    def sub(self, num):
        self.result -= num
        return self.result
                
    
    
```

```python
cal1 = Calculator()
cal2 = Calculator()
```

```python
print(cal1.add(3))


print(cal2.sub(3))
```
