```python
codes = ["TECH", "FIN", "HEALTH"]
codes.sort() # 오름차순 정렬
print (codes) # ['FIN', 'HEALTH', 'TECH']
codes.sort(reverse=True) # 내림차순 정렬
print(codes) # ['TECH', 'HEALTH', 'FIN']    

#sort는 return값이 없다. 출력시 codes만 출력해야함.
```

```python
codes = ["TECH", "FIN", "HEALTH"]
codes.reverse()
print("종목 코드 사전역순:", codes) # ['HEALTH', 'FIN', 'TECH']
#reverse는 return값이 없다. 출력시 codes만 출력해야함.
```

### 튜플 자료형 ("tuple"은 소괄호로 표현)
 * 순서가 있고, 수정/삭제 불가능, 고정된 거래 식별 정보에 활용된다
 * 요소가 하나면 (값,)

```python
trade = ("2026-01-05","ETF","BUY",3)
# 거래 날짜 : 2026-01-05 수량 3
print("거래날짜:",trade[0], "수량:",trade[3])
print("거래 정보 일부:",trade[1:])
print("자료형:",type(trade))
print("요소 개수:", len(trade))
print("한 요소 튜플:",('ETF',),type(('ETF',)))
```

```python
trade = ("2026-01-05",'ETF','BUY',3)
try:
    del trade[0] # 튜플 요소 삭제 불가
except TypeError as error:
    print("요소 삭제 불가 : ",type(error).__name__)
```

```python
try:
    trade[0] = "2026-01-06" #튜플 요소 변경도 불가
except TypeError as error:
    print("요소 변경 불가:",type(error).__name__)
```

```python
trade_id = ('2026-01-06','ETF')
quantity_price = (3,50000)


print("튜플 연결:", trade_id + quantity_price)
print("튜플 반복:", quantity_price * 3)
print("튜플 길이:", len(trade_id))


```

* 인덱스가 아니라 키로 조회

```python
price_by_code = {'ETF':50000,'BOND':102000}
quantity_by_id = {101:3}
price_history = {'ETF':[49000,50000,50500]} #리스트를 값으로 사용 가능
price_by_day = {('ETF','2026-01-05'):50000}
```

```python
holding = {'ETF':3}
holding['BOND'] = 2
holding['ETF'] = 4
holding['CASH'] = [100000,50000]
print("추가/수정:", holding, "자료형:", type(holding))
```

### keys(), values(), items(), get(), in
* keys(), values(), items()는 동적인 뷰(view)
* get()는 키가 없으면 기본적으로 None 반환, [...]는 키가 없으면 KeyError 발생

```python
portfolio = {'ETF': 3, 'BOND': 2, 'CASH': 100000}
print("키 뷰:", portfolio.keys())
print("키 목록:", list(portfolio.keys()))
print("값 뷰:", portfolio.values())
print("항목 뷰:", portfolio.items())
print("항목 목록:", list(portfolio.items()))
```

```python
portfolio = {'ETF': 3, 'BOND': 2, 'CASH': 100000}

for code in portfolio.keys():
    print("종목/자산:", code)
    
for data in portfolio.items():
    print("data:",data)
    
for code, amount in portfolio.items():
    print("키:",code, "값:",amount)
```

```python
portfolio = {'ETF': 3, 'BOND': 2, 'CASH': 100000}
print('ETF 조회:', portfolio.get("ETF"),portfolio["ETF"])
print("없는 키 기본 값:",portfolio.get("GOLD",0),portfolio.get("GOLD"))

try: 
    print(portfolio['GOLD'])
except KeyError as error:
    print("대괄호로 없는 키 조회:", type(error).__name__)
```
