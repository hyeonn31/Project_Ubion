### keys(), values(), items(), get(), in
* keys(), values(), items()는 동적인 뷰(view)
* get()는 키가 없으면 기본적으로 None 반환, [...]는 키가 없으면 KeyError 발생

```python
portfolio = {'ETF': 3, 'BOND': 2, 'CASH': 100000}
print('키 뷰:', portfolio.keys())
print('키 목록:', list(portfolio.keys()))
print('값 뷰:', portfolio.values())
print('항목 뷰:', portfolio.items())
print('항목 목록:', list(portfolio.items()))
```

```python
portfolio = {'ETF': 3, 'BOND': 2, 'CASH': 100000}

for code in portfolio.keys():
    print('종목/자산:', code)

for data in portfolio.items():
    print('data:', data)
      
for code, amount in portfolio.items():
    print('키:', code, '값:', amount)
```

```python
portfolio = {'ETF': 3, 'BOND': 2, 'CASH': 100000}
print('ETF 조회:', portfolio.get('ETF'), portfolio['ETF'])
print('없는 키 기본값:', portfolio.get('GOLD'))  # 값이 없으면 None 
print('없는 키에 기본값 지정:', portfolio.get('GOLD', 0))
```

```python
try:
    print(portfolio['GOLD'])
except KeyError as error:
    print('대괄호로 없는 키 조회:', type(error).__name__)
```

```python
print('ETF 키 존재?:', 'ETF' in portfolio)
```

### 교집합, 합집합, 차집합

```python
account_a = {'ETF', 'BOND', 'CASH', 'GOLD'}
account_b = {'ETF', 'GOLD', 'REIT'}

print('공통 자산(&):', sorted(account_a & account_b))
print('공통 자산(intersection):', sorted(account_a.intersection(account_b)))
print('전체 자산(|):', sorted(account_a | account_b))
print('전체 자산(union):', sorted(account_a.union(account_b)))
print('A에만 있는 자산(-):', sorted(account_a - account_b))
print('A에만 있는 자산(difference):', sorted(account_a.difference(account_b)))
```

### 원소 추가·여러 원소 추가·원소 삭제

```python
watchlist = set(['ETF', 'BOND', 'CASH'])
watchlist.add('GOLD')
print('add:', sorted(watchlist))
watchlist.update(['REIT', 'FX'])
print('update:', sorted(watchlist))
watchlist.remove('BOND')
print('remove:', sorted(watchlist))
```

## 변수, 참조와 복사

```python
prices = [70000, 71000, 69000]
alias = prices

print('내용이 같은가?:', prices == alias)
print('식별자가 같은가?:', id(prices) == id(alias))
print('같은 객체인가?:', prices is alias)

print('변경 전:', alias)
prices[1] = 72000
print('변경 후 원본/별칭:', prices, alias)
```

```python
id(prices), id(alias)
```

## 서로 다른 리스트 만들기: 슬라이스, copy.copy(), list.copy()


```python
from copy import copy

prices = [70000, 71000, 69000]

by_slice = prices[:]
by_module = copy(prices)
by_method = prices.copy()

prices[1] = 72000
print('원본:', prices)
print('슬라이스:', by_slice)
print('copy.copy:', by_module, '같은 객체?:', by_module is prices)
print('list.copy:', by_method, '같은 객체?:', by_method is prices)
```

```python
account_a = {'ETF','BOND','CASH'}
account_b = {'ETF','REIT'}

# 1. 교집합: 두 계좌 모두에 있는 자산
print("교집합:", account_a & account_b)    # {'ETF'}

# 2. 합집합: 둘 중 하나라도 가진 자산
print("합집합:", account_a | account_b)    # {'ETF', 'BOND', 'CASH', 'REIT'}

# 3. A만의 자산: A에는 있고 B에는 없는 것
print("A만:", account_a - account_b)       # {'BOND', 'CASH'}
```

```python
# 5. 빈 보유 리스트의 참/거짓을 확인하고, 리스트를 별칭으로 대입할 때와(=) 복사할 때 원본의 변화가 전달되는지 확인하세요

# 1. 빈 리스트의 참/거짓
holdings = []
print("빈 리스트:", bool(holdings))        # False

if not holdings:
    print("보유 자산이 없습니다")            # 출력됨

# 2. 별칭(=) vs 복사
original = ['ETF', 'BOND']

alias = original           # 별칭: 같은 리스트를 가리킴
copied = original.copy()   # 복사: 새 리스트를 만듦 (original[:]도 같음)

original.append('CASH')    # 원본 변경

print("원본:", original)   # ['ETF', 'BOND', 'CASH']
print("별칭:", alias)      # ['ETF', 'BOND', 'CASH']  → 변화가 전달됨
print("복사:", copied)     # ['ETF', 'BOND']          → 변화가 전달 안 됨

print(alias is original)   # True   (같은 객체)
print(copied is original)  # False  (다른 객체)


```
