```python
f= open("새파일.txt", "w", encoding="utf-8")
for i in range(1,11):
    data = f"{i}번째 줄입니다.\n"
    f.write(data)
f.close()
```

```python
# readline 함수 : 한 줄만 읽는 함수

f = open("새파일.txt", "r")
line = f.readline()
print(line)
f.close()
```

```python
# readlines 함수 다 읽는 함수

f = open("새파일.txt", "r")
lines = f.readlines()
print(lines)
```

```python
for line in lines:
    print(line.strip())
```

```python
# with 문과 함께 사용
with open("foo,txt","w") as f:
    f.write("you need python")
```

```python
# 내용 추가
f = open("새파일.txt", "a")
for i in range(11,21):
    data = "%d번째 줄입니다.\n" %i
    f.write(data)
f.close()
```
