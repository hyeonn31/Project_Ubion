```python
import re
from pypdf import PdfReader
import nbformat as nbf

pdf = "ADP 파이썬 판다스 실습 문제 50제 (문제지).pdf"
text = "\n".join(page.extract_text() for page in PdfReader(pdf).pages)

# 문제 번호(예: "1." / "Q1" / "문제 1")를 기준으로 자르기 — PDF 형식에 맞게 정규식 조정 필요
problems = re.split(r"\n(?=\d{1,2}\.\s)", text)

nb = nbf.v4.new_notebook()
nb.cells.append(nbf.v4.new_code_cell("import pandas as pd\nimport numpy as np"))
for p in problems:
    nb.cells.append(nbf.v4.new_markdown_cell(p.strip()))
    nb.cells.append(nbf.v4.new_code_cell("# 풀이\n"))

nbf.write(nb, "ADP_판다스_50제_풀이.ipynb")
```
