### 데이터 인코딩

* 레이블 인코딩(Label encoding)

```python
from sklearn.preprocessing import LabelEncoder

items=['TV', '냉장고', '전자렌지', '컴퓨터', '선풍기', '믹서', '믹서']

encoder = LabelEncoder()
encoder.fit(items)  # 중복 제거된 데이터 추출. 전달한 데이터에서 어떤 항목들이 있는지 살펴보고, 변환 기준을 학습해요. 중복 항목은 한 종류로 취급
```

```python
# 중복된 데이터 확인
print(encoder.classes_) # classes_  에 할당되어 저장된다
```

```python
labels = encoder.transform(items)   # 숫자로 인코딩 보면 labels에 다시 할당된 값은 없다.
print('인코딩 변환값:', labels)
```

```python
print('디코딩 원본 값:', encoder.inverse_transform([4,5,2,0,1,1,3,3]))
```

* 원-핫 인코딩(One-Hot encoding)

```python
from sklearn.preprocessing import LabelEncoder
import numpy as np

items=['TV', '냉장고', '전자렌지', '컴퓨터', '선풍기', '믹서', '믹서']

# 숫자값으로 변환을 위해 LabelEncoder로 변환
encoder = LabelEncoder()
encoder.fit(items)  # 중복 제거된 데이터 추출
labels = encoder.transform(items)
labels
```

```python
# 2차원 데이터로 변환
labels = labels.reshape(-1,1)       # 행은 상관없이(-1) 하나의 요소로(1) 받겠다. 2로 실행도 해볼것 
labels
```

```python
# 원-핫 인코딩을 적용
from sklearn.preprocessing import OneHotEncoder
oh_encoder = OneHotEncoder()
oh_encoder.fit(labels)
oh_labels = oh_encoder.transform(labels)
oh_labels
# 압축을 했다. 라고 나오는듯
```

```python
print(oh_labels)
print('원-핫 인코딩 데이터')
print(oh_labels.toarray())          # 표를 보면서 이해하면 이해하기 편하다. 예를들어 (0,0) 칸이 1이고 (2,4) 칸이 1이다. 라는 뜻
print('원-핫 인코딩 데이터 차원')
print(oh_labels.shape)              # 이게 표를 보여준다.
```

### 피처 스케일링과 정규화
* 피처(feature)마다 값의 크기나 범위가 다를 떄 비슷한 스케일로 맞춰주는 것

* StandardScaler(표준화)
    * 평균을 0, 표준편차를 1로 맞추는 방법.

```python
from sklearn.datasets import load_iris  # 붓꽃데이터
import pandas as pd

iris= load_iris()
type(iris)
iris_data = iris.data   # 꽃잎, 꽃받침 넓이 길이
iris_data   # 이런 실 데이터를 가지고 y값(여기서는 꽃종류인것 같다)을 분류해본다. 실데이터가 = 피처
```

```python
iris_df = pd.DataFrame(data=iris_data, columns=iris.feature_names)
iris_df
```

```python
# Stan
from sklearn.preprocessing import StandardScaler

# StandardScaler 객체 생성
scaler = StandardScaler()
# fit : 각 피처의 평균과 표준편차를 계산(학습)
scaler.fit(iris_df)
# transform : 계산한 평균, 표준편차로 데이터를 변환 -> 결과는 넘파이 배열(ndarray)
iris_scaled = scaler.transform(iris_df)

# 변환된 배열을 다시 데이터프레임으로 변환
iris_df_scaled = pd.DataFrame(data=iris_scaled, columns=iris.feature_names)

print('feature 들의 평균 값')
print(iris_df_scaled.mean())
print('\nfeature 들의 분산 값')
print(iris_df_scaled.var())
```

* MinMaxScaler(정규화)

```python
from sklearn.preprocessing import MinMaxScaler

# MinMaxScaler 객체 생성 : 각 피처의 값을 0~1 범위로 변환
scaler = MinMaxScaler()
# fit : 각 피처의 최소값, 최대값을 계산
scaler.fit(iris_df)
# transform : (값 - 최소값) / (최대값 - 최소값) 으로 변환 -> 결과는 넘파이 배열(ndarray)
iris_scaled = scaler.transform(iris_df)

# 변환된 배열을 다시 데이터프레임으로 변환
iris_df_scaled = pd.DataFrame(data=iris_scaled, columns=iris.feature_names)

print('feature 들의 최소 값')
print(iris_df_scaled.min())
print('\nfeature 들의 최대 값')
print(iris_df_scaled.max())
```

* Scaler를 이용하여 학습 데이터와 테스트 데이터에 fit(), transform(), fit_transform() 적용 시 유의 사항

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

train_array = np.arange(0,11).reshape(-1,1)
test_array = np.arange(0,6).reshape(-1,1)

print(train_array)
print(test_array)
```

```python
# 올바른 방법 : 학습 데이터로만 fit() 하고, 테스트 데이터는 transform()만 적용
scaler = MinMaxScaler()

# 학습 데이터(0~10)로 최소값, 최대값을 계산(학습)
scaler.fit(train_array)
train_scaled = scaler.transform(train_array)

print('원본 train_array 데이터:', np.round(train_array.reshape(-1), 2))
print('Scale된 train_array 데이터:', np.round(train_scaled.reshape(-1), 2))

# 테스트 데이터에는 fit()을 다시 하지 않고, 학습 데이터 기준 그대로 transform()만 수행
test_scaled = scaler.transform(test_array)

print('\n원본 test_array 데이터:', np.round(test_array.reshape(-1), 2))
print('Scale된 test_array 데이터:', np.round(test_scaled.reshape(-1), 2))
```
