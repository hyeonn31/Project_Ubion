## 시리즈

* dict to series

```python
# pandas 라이브러리를 pd 라는 별명으로 불러온다 (관례적으로 항상 pd 사용)
import pandas as pd

# 딕셔너리 : {키: 값} 형태의 자료형
dict_data = {'a': 1, 'b': 2, 'c': 3}

# 딕셔너리 → 시리즈로 변환
#   키(a, b, c)  → 시리즈의 '인덱스(이름표)'가 됨
#   값(1, 2, 3)  → 시리즈의 '데이터 값'이 됨
sr = pd.Series(dict_data)

print(type(sr))   # <class 'pandas.core.series.Series'> → 시리즈 객체인지 확인
print()           # 빈 줄 출력(보기 좋게 띄우기용)
print(sr)         # 왼쪽 = 인덱스, 오른쪽 = 값, 마지막 줄 dtype = 값의 자료형
```

* list to series - series index

```python
import pandas as pd

# 리스트 : 순서가 있는 자료형. 문자열, 실수, 정수, 불리언 등 섞여 있어도 된다
list_data = ['2019-01-02', 3.14, 'ABC', 100, True]

# 리스트 → 시리즈로 변환
#   리스트에는 '키'가 없으므로 인덱스가 0, 1, 2, 3, 4 로 자동 생성된다
sr = pd.Series(list_data)

# 여러 자료형이 섞여 있으므로 dtype 이 object(=아무거나) 로 표시된다
print(sr)
```

```python
# 시리즈는 '인덱스'와 '값' 두 부분으로 구성된다. 각각 따로 꺼낼 수 있다
idx = sr.index    # 인덱스만 꺼내기 → RangeIndex(start=0, stop=5, step=1) : 0~4 까지라는 뜻
val = sr.values   # 값만 꺼내기 → numpy 배열(array) 형태로 반환됨

print(idx)
print('\n')       # '\n' = 줄바꿈 문자. print 자체 줄바꿈 + \n 이라 빈 줄이 2개 생긴다
print(val)
```

* tuple to series

```python
import pandas as pd

# 튜플 : 리스트와 비슷하지만 ( ) 로 만들고, 수정할 수 없는 자료형
tup_data = ('영인', '2010-05-01', '여', True)

# 튜플 → 시리즈로 변환
#   index=[...] 옵션으로 인덱스 이름을 직접 지정 (0,1,2,3 대신 이름표를 붙임)
#   ※ index 개수 = 데이터 개수 (여기선 4개) 가 반드시 같아야 한다
sr = pd.Series(tup_data, index=['이름', '생년월일', '성별', '학생여부'])
print(sr)
```

```python
# 원소를 1개 선택
# 방법 1) iloc[정수] : '위치(순서)' 로 선택. 0번째 = 첫 번째 원소 → '영인'
print(sr.iloc[0])

# 방법 2) [이름] : '인덱스 이름' 으로 선택 → '영인'
#   (sr.loc['이름'] 과 같은 의미)
print(sr['이름'])
print('\n')
```

```python
# 여러 개의 원소를 선택

# 방법 1) 슬라이싱 iloc[시작:끝] : 위치 1 ~ 2 까지 선택
#   ※ 끝 번호(3)는 포함되지 않는다! → 1번(생년월일), 2번(성별) 만 선택
print(sr.iloc[1 : 3])
print('\n')

# 방법 2) 인덱스 이름을 '리스트'로 전달 → 여러 개 선택
#   대괄호가 2겹 [[ ]] 인 이유 : 바깥 [] = 선택 문법, 안쪽 [] = 이름 목록(리스트)
print(sr[['생년월일', '성별']])
```

## 데이터프레임

* dict to dataframe

```python
import pandas as pd

# 데이터프레임 = 행(row)과 열(column)로 이루어진 '표(엑셀 시트)' 형태
# 딕셔너리로 만들 때:
#   키('c0' ~ 'c4')       → 열 이름
#   값(리스트 [1,2,3] 등)  → 그 열에 위에서부터 들어갈 데이터
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}

# 딕셔너리 → 데이터프레임 변환 (행 인덱스는 0, 1, 2 자동 생성)
df = pd.DataFrame(dict_data)

print(type(df))   # <class 'pandas.core.frame.DataFrame'>
print('\n')
print(df)         # 5개 열(c0~c4) × 3개 행(0~2) 짜리 표가 출력됨
```

* list to dataframe

```python
import pandas as pd

# 2차원 리스트(리스트 안의 리스트)로 데이터프레임 만들기
#   안쪽 리스트 하나 = '행' 하나  → [15, '남', '덕영중'] 이 첫 번째 행
#   index=[...]   : 행 이름 지정 (행 개수와 같아야 함 → 2개)
#   columns=[...] : 열 이름 지정 (열 개수와 같아야 함 → 3개)
df = pd.DataFrame([[15, '남', '덕영중'], [17, '여', '수리중']],
                   index=['준서', '예은'],
                   columns=['나이', '성별', '학교'])

print(df)            # 데이터프레임 전체 출력
print('\n')
print(df.index)      # 행 인덱스(행 이름)만 출력 → Index(['준서', '예은'])
print('\n')
print(df.columns)    # 열 이름만 출력 → Index(['나이', '성별', '학교'])
print('\n')
```

* 열, 행 이름 변경

```python
import pandas as pd

df = pd.DataFrame([[15, '남', '덕영중'], [17, '여', '수리중']],
                   index=['준서', '예은'],
                   columns=['나이', '성별', '학교'])

print(df)            # 변경 전 데이터프레임
print('\n')

# rename() : 이름 바꾸기. {기존이름: 새이름} 딕셔너리로 전달
# [방법 1] inplace=True → 원본 df 자체를 바로 수정 (반환값 없음, None)
df.rename(columns={'나이':'연령', '성별':'남녀', '학교':'소속'}, inplace=True)

# [방법 2] inplace 생략(기본 False) → 원본은 그대로, '수정된 새 데이터프레임'을 반환
#   그래서 df = ... 로 다시 담아줘야 변경이 반영된다
df = df.rename(index={'준서':'학생1', '예은':'학생2' })

print(df)            # 열 이름, 행 이름이 모두 바뀐 결과
```

* remove row

```python
import pandas as pd

# 시험 점수 데이터 : 키 = 과목(열 이름), 값 = 학생 3명의 점수
exam_data = {'수학' : [ 90, 80, 70], '영어' : [ 98, 89, 95],
             '음악' : [ 85, 95, 100], '체육' : [ 100, 90, 90]}

# index 로 행 이름(학생 이름)을 지정
df = pd.DataFrame(exam_data, index=['서준', '우현', '인아'])
print(df)
print('\n')
```

```python
# df[:] : 전체 행을 슬라이싱 → df 와 같은 내용의 데이터프레임을 얻음
#   ※ 원본을 안전하게 보존하려면 df.copy() 사용을 권장 (아래 셀 참고)
df2 = df[:]

# drop('행이름') : 해당 행 삭제. axis 를 안 쓰면 기본값 axis=0(행 방향)
#   inplace=True → df2 자체에서 바로 삭제
df2.drop('우현', inplace=True)
print(df2)           # '우현' 행이 사라진 결과
```

```python
# 우현, 인아 삭제
df3 = df[:]

# 여러 행을 지울 때는 행 이름을 '리스트'로 전달
#   axis=0 : 행(row)을 삭제하라는 뜻 (기본값이라 생략 가능)
df3.drop(['우현', '인아'], axis=0, inplace=True)
print(df3)           # '서준' 행만 남음
```

* remove column

```python
# df.copy() : 원본과 완전히 분리된 '복사본' 생성
#   → df4 를 수정해도 원본 df 는 절대 바뀌지 않는다 (가장 안전한 복사 방법)
df4 = df.copy()  # [:]

# axis=1 : 열(column)을 삭제하라는 뜻 → '수학' 열 삭제
df4.drop('수학', axis=1, inplace=True)

# 셀의 마지막 줄에 변수만 쓰면 print 없이도 결과가 표(HTML) 형태로 예쁘게 출력됨
df4
```

```python
df5 = df.copy()

# 여러 열 삭제 : 열 이름을 리스트로 전달 + axis=1
df5.drop(['영어', '음악'], axis=1, inplace=True)
print(df5)           # 수학, 체육 열만 남음
```

* select row(행)

```python
import pandas as pd

# 행 선택 실습을 위해 데이터프레임을 다시 만든다
# (위에서 drop 등을 실행했을 수 있으므로 깨끗한 상태로 초기화)
exam_data = {'수학' : [ 90, 80, 70], '영어' : [ 98, 89, 95],
             '음악' : [ 85, 95, 100], '체육' : [ 100, 90, 90]}

df = pd.DataFrame(exam_data, index=['서준', '우현', '인아'])
print(df)
print('\n')
```

```python
# 행 인덱스를 사용하여 행 1개를 선택 : loc(문자 인덱스), iloc(숫자 인덱스)
#   loc  = location         → '이름(라벨)' 으로 찾기
#   iloc = integer location → '위치 번호(0부터)' 로 찾기

# 서준 행 검색 : loc 인덱서 활용 (행 이름 '서준' 으로 찾기)
label1 = df.loc['서준']

# 0번째 행 검색 : iloc 인덱서 활용 (0번째 = 첫 번째 행 = 서준)
position1 = df.iloc[0]

# 둘 다 같은 행(서준)을 가리킨다
# 행 1개만 뽑으면 결과는 '시리즈' (열 이름이 인덱스가 되어 세로로 출력됨)
print(label1)
print('\n')
print(position1)
```

```python
# 행 인덱스를 사용하여 2개 이상의 행 선택
# 서준, 우현 : 인덱싱(loc, iloc 사용)
#   원하는 행을 '리스트'로 전달 → 대괄호 2겹 [[ ]]
#   행을 2개 이상 뽑으면 결과는 '데이터프레임' (표 형태)
label2 = df.loc[['서준', '우현']]   # 이름으로 선택
position2 = df.iloc[[0, 1]]         # 위치 번호로 선택
print(label2)
print('\n')
print(position2)
print('\n')

# 서준, 우현 : 슬라이싱(loc, iloc 사용)
#   ★ 매우 중요한 차이 ★
#   loc['서준':'우현'] → 끝('우현')을 '포함'한다   (서준, 우현)
#   iloc[0:2]          → 끝(2)을 '포함하지 않는다' (0, 1번 = 서준, 우현)
label3 = df.loc['서준':'우현']
position3 = df.iloc[0:2]
print(label3)
print('\n')
print(position3)
```
