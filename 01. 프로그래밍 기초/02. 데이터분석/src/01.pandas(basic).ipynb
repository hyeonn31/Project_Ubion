## 시리즈

* dict to series

```python
import pandas as pd

dict_data = {'a': 1, 'b': 2, 'c': 3}

sr = pd.Series(dict_data)
print(type(sr))
print()
print(sr)
```

* list to series - series index

```python
import pandas as pd

list_data = ['2019-01-02', 3.14, 'ABC', 100, True]
sr = pd.Series(list_data)

print(sr)
```

```python
idx = sr.index
val = sr.values
print(idx)
print('\n')
print(val)
```

* tuple to series

```python
import pandas as pd

tup_data = ('영인', '2010-05-01', '여', True)
sr = pd.Series(tup_data, index=['이름', '생년월일', '성별', '학생여부'])
print(sr)
```

```python
# 원소를 1개 선택
print(sr.iloc[0])  
print(sr['이름'])
print('\n')
```

```python
# 여러 개의 원소를 선택 
print(sr.iloc[1 : 3])  
print('\n')
print(sr[['생년월일', '성별']])
```

## 데이터프레임

* dict to dataframe

```python
import pandas as pd

dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}

df = pd.DataFrame(dict_data)
print(type(df)) 
print('\n')
print(df)
```

* list to dataframe

```python
import pandas as pd

df = pd.DataFrame([[15, '남', '덕영중'], [17, '여', '수리중']], 
                   index=['준서', '예은'],
                   columns=['나이', '성별', '학교'])

print(df)            #데이터프레임
print('\n')
print(df.index)      #행 인덱스
print('\n')
print(df.columns)    #열 이름
print('\n')
```

* 열, 행 이름 변경

```python
import pandas as pd

df = pd.DataFrame([[15, '남', '덕영중'], [17, '여', '수리중']], 
                   index=['준서', '예은'],
                   columns=['나이', '성별', '학교'])

print(df)            #데이터프레임
print('\n')

df.rename(columns={'나이':'연령', '성별':'남녀', '학교':'소속'}, inplace=True)
df = df.rename(index={'준서':'학생1', '예은':'학생2' })

print(df)
```

* remove row

```python
import pandas as pd

exam_data = {'수학' : [ 90, 80, 70], '영어' : [ 98, 89, 95],
             '음악' : [ 85, 95, 100], '체육' : [ 100, 90, 90]}

df = pd.DataFrame(exam_data, index=['서준', '우현', '인아'])
print(df)
print('\n')
```

```python
df2 = df[:]
df2.drop('우현', inplace=True)
print(df2)
```

```python
# 우현, 인아 삭제
df3 = df[:]
df3.drop(['우현', '인아'], axis=0, inplace=True)  # axis=0 : default값(생략 가능)
print(df3)
```

* remove column

```python
df4 = df.copy()  # [:]
df4.drop('수학', axis=1, inplace=True)
df4
```

```python
df5 = df.copy()
df5.drop(['영어', '음악'], axis=1, inplace=True)
print(df5)
```

* select row(행)

```python
import pandas as pd

exam_data = {'수학' : [ 90, 80, 70], '영어' : [ 98, 89, 95],
             '음악' : [ 85, 95, 100], '체육' : [ 100, 90, 90]}

df = pd.DataFrame(exam_data, index=['서준', '우현', '인아'])
print(df)
print('\n')
```

```python
# 행 인덱스를 사용하여 행 1개를 선택 : loc(문자 인덱스), iloc(숫자 인덱스)
# 서준 행 검색 : loc 인덱서 활용
label1 = df.loc['서준'] 

# 0번째 행 검색 : iloc 인덱서 활용
position1 = df.iloc[0]

print(label1)
print('\n')
print(position1)
```

```python
# 행 인덱스를 사용하여 2개 이상의 행 선택
# 서준, 우현 : 인덱싱(loc, iloc 사용)
label2 = df.loc[['서준', '우현']]
position2 = df.iloc[[0, 1]]
print(label2)
print('\n')
print(position2)
print('\n')

# 서준, 우현 : 슬라이싱(loc, iloc 사용)
label3 = df.loc['서준':'우현']
position3 = df.iloc[0:2]
print(label3)
print('\n')
print(position3)
```
