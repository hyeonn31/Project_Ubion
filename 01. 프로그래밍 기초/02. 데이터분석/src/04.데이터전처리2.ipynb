## 1.누락 데이터 처리(NaN : Not a Number, numpy.nan)
* 누락데이터가 많으면 데이터 품질 저하, AI 알고리즘이 왜곡되는 현상 생김

* 누락 데이터 확인 

```python
import seaborn as sns

# seaborn 내장 타이타닉 데이터셋 불러오기 (891행 x 15열)
df = sns.load_dataset('titanic')

# deck 열의 값별 개수 집계
# dropna=False : NaN 값까지 집계
nan_deck = df['deck'].value_counts(dropna=False)
print(nan_deck)
```

```python
# isnull() 메서드로 누락 데이터 찾기
# 누락 데이터면 True, 유효한 데이터가 존재하면 False
print(df.head().isnull())      # head() : 앞 5행만 확인
```

```python
# notnull() 메서드로 누락 데이터 찾기
# 유효한 데이터가 존재하면 True, 누락 데이터면 False
print(df.head().notnull())
```

```python
# isnull() 메서드로 누락 데이터 개수 구하기
# True = 1, False = 0 으로 계산되므로 sum()하면 NaN 개수가 됨
# axis=0 : 행 방향(위->아래)으로 더함 -> 열별 NaN 개수
print(df.head().isnull().sum(axis=0))
```

```python
# 전체 데이터의 열별 누락 데이터 개수
df.isnull().sum(axis=0)
```

```python
import seaborn as sns

df = sns.load_dataset('titanic')
missing_df = df.isnull()        # 전체 데이터를 True(NaN) / False 로 변환

# 열마다 NaN(True) 개수 출력
for col in missing_df.columns:
    missing_count = missing_df[col].value_counts()    # 각 열의 NaN 개수 파악 (True, False 각각의 개수)
    # print(missing_count)

    try:
        print(col, ': ', missing_count[True])   # NaN 값이 있으면 개수를 출력
    except:
        print(col, ': ', 0)                     # NaN 값이 없으면(True가 없어서 KeyError) 0개 출력

# NaN이 아닌 값이 500개 이상인 열을 남김 - deck 열(891개 중 688개의 NaN 값)
# axis=1 : 열 단위로 삭제, thresh=500 : 유효한 값이 500개 미만인 열은 삭제
df_thresh = df.dropna(axis=1, thresh=500)
print(df_thresh.columns)        # deck 열이 삭제된 것 확인
```

```python
# NaN이 500개 이하인 열만 남기고 싶을 경우
# df.isnull().sum() <= 500 : 열별로 True/False 를 가진 시리즈(mask)
# df.loc[:, mask] : 모든 행, mask가 True인 열만 선택
df.loc[:, df.isnull().sum() <= 500]
```

```python
# age 열에 나이 데이터가 없는 모든 행을 삭제 - age 열(891개 중 177개의 NaN 값)
# subset=['age'] : NaN을 검사할 열 지정. 여러개의 변수가 올 경우, how='any'에 의해 한개의 변수라도 NaN이면 행을 삭제해라
# how='all'이면 모두 NaN일 경우에만 삭제
# axis=0 : 행 단위로 삭제
df_age = df.dropna(subset=['age'], how='any', axis=0)
df_age.head()
```

## 3.데이터 표준화

#### 3-1.단위 환산

```python
# 측정 단위를 동일하게 맞추기
    # 마일, 야드, 온스(영미권) -> 미터 , 평, 그램(한국)

'''
UCI 자동차 연비 데이터셋
    mpg(갤런당 마일 : mile per gallon) : kpl(단위 -> 리터당 킬로미터(km/L))
    1마일 = 1.60934km, 1갤런 = 3.78541리터, 1mpg = 0.425km/L
'''
```

```python
import pandas as pd

# header=None : CSV 첫 줄이 열 이름이 아니라 데이터이므로 열 이름 없이 읽음
df = pd.read_csv("./data/auto-mpg.csv", header=None)

# 열 이름을 지정
df.columns = [
    "mpg",           # 연비 (갤런당 마일)
    "cylinders",     # 실린더 수
    "displacement",  # 배기량
    "horsepower",    # 출력 (마력)
    "weight",        # 차중
    "acceleration",  # 가속능력
    "model year",    # 출시 년도
    "origin",        # 제조국 (1:USA, 2:EU, 3:JAPAN)
    "name",          # 모델명
]

print(df.head(3))
```

```python
# mpg -> kpl 환산 계수 (1마일 km / 1갤런 L)
mpg_to_kpl = 1.60934 / 3.78541
df["kpl"] = df["mpg"] * mpg_to_kpl      # 새 열 kpl 추가
print(df.head(3))
```

```python
# kpl 열을 소수점 둘째 자리까지 반올림
df["kpl"] = df["kpl"].round(2)
print(df.head(3))
```

##  4.범주형(카테고리) 데이터 처리

#### 4-1.구간 분할
* 연속 변수를 일정 구간으로 나누고 카테고리컬 데이터(범주형 이산 데이터)로 변환하는 것

```python
import numpy as np
import pandas as pd

df = pd.read_csv("./data/auto-mpg.csv", header=None)
# 열 이름을 지정
df.columns = [
    "mpg",           # 연비 (갤런당 마일)
    "cylinders",     # 실린더 수
    "displacement",  # 배기량
    "horsepower",    # 출력 (마력)
    "weight",        # 차중
    "acceleration",  # 가속능력
    "model year",    # 출시 년도
    "origin",        # 제조국 (1:USA, 2:EU, 3:JAPAN)
    "name",          # 모델명
]

# 누락 데이터('?') 삭제
df['horsepower'] = df['horsepower'].replace('?', np.nan)      # '?'을 np.nan으로 변경
df.dropna(subset=['horsepower'], axis=0, inplace=True)   # 누락데이터 행을 삭제
df['horsepower'] = df['horsepower'].astype('float')      # 문자열을 실수형으로 변환

# horsepower 열의 자료형 확인
print(df['horsepower'].dtypes)  
```

```python
# np.histogram 함수로 3개의 bin으로 나누는 경계 값의 리스트 구하기
# count : 각 구간의 데이터 개수, bin_dividers : 구간 경계값(4개)
# bins=3 : 최소값~최대값을 같은 간격으로 3등분

count, bin_dividers = np.histogram(df['horsepower'], bins=3)
print(bin_dividers)
print(count)

'''
1구간 : 46.<=  horsepower < 107.33333333
2구간 : 107.33333333 <=  horsepower < 168.66666667
3구간 : 168.66666667 <=  horsepower <= 230.
'''
```

```python
# 3개의 bin에 이름 지정
bin_names = ['저출력', '보통출력', '고출력']

# pd.cut() : 연속된 숫자 데이터를 구간별 범주로 나눔
df['hp_bin'] = pd.cut(x=df['horsepower'],     # 데이터 배열
                    bins=bin_dividers,      # 경계 값 리스트
                    labels=bin_names,       # bin 이름
                    include_lowest=True)    # 첫 경계값(46.0)도 1구간에 포함

print(df[['horsepower', 'hp_bin']].head(15))
```

## 5.정규화(Normalization)
* 0~1, -1~1 까지의 범위를 가지고 있음

```python
import numpy as np
import pandas as pd

# 데이터를 새로 불러옴 (앞 셀에서 df가 바뀌었을 수 있으므로)
df = pd.read_csv("./data/auto-mpg.csv", header=None)
# 열 이름을 지정
df.columns = [
    "mpg",           # 연비 (갤런당 마일)
    "cylinders",     # 실린더 수
    "displacement",  # 배기량
    "horsepower",    # 출력 (마력)
    "weight",        # 차중
    "acceleration",  # 가속능력
    "model year",    # 출시 년도
    "origin",        # 제조국 (1:USA, 2:EU, 3:JAPAN)
    "name",          # 모델명
]

# 누락 데이터('?') 삭제
df['horsepower'] = df['horsepower'].replace('?', np.nan)      # '?'을 np.nan으로 변경
df.dropna(subset=['horsepower'], axis=0, inplace=True)   # 누락데이터 행을 삭제
df['horsepower'] = df['horsepower'].astype('float')      # 문자열을 실수형으로 변환
```

```python
# 정규화 전 horsepower 통계 확인 (min 46 ~ max 230)
print(df.horsepower.describe())
```

```python
# 정규화 : 각 열의 값을 그 열의 최대값(절대값)으로 나누어 0 ~ 1 범위로 변환
# 열마다 숫자 크기(단위)가 달라도 같은 범위로 맞춰서 비교 가능하게 함
df.horsepower = df.horsepower / abs(df.horsepower.max()) # abs : 절대값

print(df.horsepower.head())
print('\n')
print(df.horsepower.describe())     # max가 1.0인 것 확인
```

```python
import pandas as pd

# 주식 거래 예시 데이터프레임 생성 (거래일은 아직 문자열)
df = pd.DataFrame({
    '거래일': [
        '2025-01-10', '2025-01-20',
        '2025-02-10', '2025-02-20',
        '2026-01-10', '2026-01-20'
    ],
    '종가': [70000, 72000, 71000, 75000, 80000, 82000],      # 장 마감 가격
    '거래량': [1000, 1500, 1200, 1800, 2000, 2500]          # 거래된 주식 수
})

# 문자열을 Timestamp로 변환
df['거래일'] = pd.to_datetime(df['거래일'])

# 월 단위 Period 생성 -> 새 열 거래월 추가
# .dt : 날짜형 열에서 날짜 관련 기능을 쓰기 위한 접근자
df['거래월'] = df['거래일'].dt.to_period('M')

print(df)
```

```python
# 월별 거래량 합계 계산
# groupby('거래월') : 같은 달끼리 묶음 -> ['거래량'].sum() : 묶음별 거래량 합계
monthly = df.groupby('거래월')['거래량'].sum()
print(monthly)
```

```python
# 월별 종가와 수익률 계산
# last() : 각 달의 마지막 거래일 종가 = 월말 종가
monthly_close = df.groupby('거래월')['종가'].last()
monthly_close
```

```python
# 전월 대비 수익률(%)
# pct_change() : (이번 값 - 이전 값) / 이전 값 -> 첫 달은 비교 대상이 없어서 NaN
monthly_return = monthly_close.pct_change() * 100

print("\n월별 수익률(%):")
print(monthly_return.round(2))      # 소수점 둘째 자리까지 반올림

# (75000 - 72000) / 72000  = 0.041667 : 이전달 종가 대비 변화율 소수로 반환
```

#### 6-2.시계열 데이터 만들기

* Timestamp 배열

```python
import pandas as pd

# Timestamp의 배열 만들기 - 월 간격, 월의 시작일 기준
ts_ms = pd.date_range(start='2019-01-01',    # 날짜 범위의 시작
                end=None,                 # 날짜 범위의 끝
                periods=6,                # 생성할 Timestamp의 개수
                freq='MS',                # 시간 간격 (MS: 월의 시작일)
                tz='Asia/Seoul')          # 시간대(timezone)
print(ts_ms)
```

```python
# .dt : 날짜형 열에서 연/월/일을 꺼내는 접근자
df['Year'] = df['new_Date'].dt.year      # 연
df['Month'] = df['new_Date'].dt.month    # 월
df['Day'] = df['new_Date'].dt.day        # 일
print(df.head())
```

* 인덱싱

```python
import pandas as pd

# 데이터를 새로 불러옴 (앞에서 Year, Month, Day 열을 추가했으므로)
df = pd.read_csv('./data/stock-data.csv')
df['new_Date'] = pd.to_datetime(df['Date'])     # 문자열 -> Timestamp(날짜형)
print(df.head())
```

```python
# new_Date 열을 인덱스로 지정 -> 날짜로 행 선택 가능 (한 번만 실행할 것)
df.set_index('new_Date', inplace=True)
# 날짜 오름차순 정렬 (원본은 최신 날짜부터 거꾸로 정렬되어 있음)
# 날짜 범위 슬라이싱(:)은 인덱스가 오름차순이어야 동작함
df.sort_index(inplace=True)
print(df.head())
```

```python
# 날짜 인덱스를 이용하여 데이터 선택 - 2018
# 연도만 써도 해당 연도의 모든 행 선택 (부분 문자열 인덱싱)
df_y = df.loc['2018']
print(df_y.head())
```

```python
# 2018-07 : 2018년 7월의 모든 행 선택
df_ym = df.loc['2018-07']
print(df_ym)
```

```python
# 2018-07, start, high
# df.loc[행, 열] : 2018년 7월 행 중에서 Start 열부터 High 열까지 선택
df_ym_cols = df.loc['2018-07', 'Start':'High']
print(df_ym_cols)
```

```python
# 2018-07-02 : 특정 날짜 하루의 데이터 선택
df_ymd = df.loc['2018-07-02']
print(df_ymd)
```

```python
# 2018-06-08~2018-06-20
# 시작:끝 날짜 범위 지정 (끝 날짜도 포함)
df_ymd_range = df.loc['2018-06-08':'2018-06-20']    # 날짜 범위 지정
print(df_ymd_range)
```
