```python
import seaborn as sns

titanic = sns.load_dataset('titanic')     # seaborn 내장 타이타닉 데이터 (891행)
df = titanic.loc[:, ['age','fare']]       # 나이, 요금 열만 선택
df['ten'] = 10                            # 모든 행에 10을 넣은 열 추가 (행 연산 실습용)
print(df.head())
```

** 시리즈 원소에 함수 매핑

```python
# 예제 1. 종가에 10% 목표 수익률 적용
def target_price(price):     # 종가 1개를 받아서
    return price * 1.10      # 10% 올린 값 반환

# 종가 시리즈에 함수 매핑 : '목표가' 변수명 추가
df['목표가'] = df['종가'].apply(target_price)
# df['목표가'] = df['종가'].map(target_price)    # 시리즈는 map 으로도 같은 결과
# 참고) 단순 계산은 df['종가'] * 1.1 이 더 빠름
df.head()
```

```python
# 예제 2. 수익률에 따라 매수·매도 신호 만들기
# if/elif 처럼 조건이 여러 개면 연산자로는 안 되니까 함수 + apply 가 편함
def trading_signal(rate):
    if rate >= 3:            # 3% 이상
        return '강한 상승'
    elif rate > 0:           # 0 초과 ~ 3% 미만
        return '상승'
    elif rate <= -1:         # -1% 이하
        return '하락'
    else:                    # -1% 초과 ~ 0% 이하
        return '약보합'

df['시장신호'] = df['수익률(%)'].apply(trading_signal)    # 숫자 → 글자 라벨
df.head()
```

```python
# 예제 3. 종가가 100,000원 이상이면 목표가 계산, 미만이면 0

# 목표가 계산 함수: 종가 대비 10% 상승
def target_price(price):
    return price * 1.10

# lambda 안에서 조건 분기 : 참이면 target_price(x), 거짓이면 0
df['목표가'] = df['종가'].apply(
    lambda x: target_price(x) if x >= 100000 else 0
)

df.head()
```

#### 1-3.데이터프레임 객체에 함수 매핑

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]
print(df.head())
```

```python
# 각 열의 NaN 찾기 - 데이터프레임 전달하면 데이터프레임을 반환
def missing_value(x):
    return x.isnull()

# 각 열의 NaN 개수 반환 - 데이터프레임 전달하면 시리즈 반환
def missing_count(x):    # True=1, False=0 으로 더해져서 개수가 됨
    return missing_value(x).sum()

# 데이터프레임의 총 NaN 개수 - 데이터프레임 전달하면 값을 반환
def total_number_missing(x):
    return missing_count(x).sum()

# pipe : 데이터프레임 전체를 함수에 한 번 넣음 → df.pipe(f) == f(df)
# 리턴 타입은 함수가 뭘 반환하느냐에 따라 달라짐
result_df = df.pipe(missing_value)
print(result_df.head())
print(type(result_df))
```

```python
# 열별 NaN 개수 → 시리즈 (age 177, fare 0)
result_series = df.pipe(missing_count)
print(result_series)
print(type(result_series))
```

```python
# 전체 NaN 개수 → 숫자 1개 (177)
result_value = df.pipe(total_number_missing)
print(result_value)
print(type(result_value))
```

#### 4-1.데이터프레임 연결(concat)
* 기존 데이터프레임의 형태를 유지하면서 이어 붙이는 형태

```python
import pandas as pd

# 데이터프레임 만들기
# df1 : 인덱스 0~3, 열 a b c  /  df2 : 인덱스 2~5, 열 a b c d  → 인덱스 2, 3 이 겹침
df1 = pd.DataFrame({'a': ['a0', 'a1', 'a2', 'a3'],
                    'b': ['b0', 'b1', 'b2', 'b3'],
                    'c': ['c0', 'c1', 'c2', 'c3']},
                    index=[0, 1, 2, 3])

df2 = pd.DataFrame({'a': ['a2', 'a3', 'a4', 'a5'],
                    'b': ['b2', 'b3', 'b4', 'b5'],
                    'c': ['c2', 'c3', 'c4', 'c5'],
                    'd': ['d2', 'd3', 'd4', 'd5']},
                    index=[2, 3, 4, 5])

print(df1, '\n')
print(df2, '\n')
```

```python
# 2개의 데이터프레임을 위 아래 행 방향으로 이어 붙이듯 연결
# df1 에 없는 d 열은 NaN, 인덱스 2, 3 이 중복으로 생김
result1 = pd.concat([df1, df2])
print(result1, '\n')
```

```python
# ignore_index=True : index 재 넘버링함 (0 ~ 7)
result2 = pd.concat([df1, df2], ignore_index=True)  # axis = 0 : 행방향
print(result2, '\n')
```

```python
# 2개의 데이터프레임을 좌우 열 방향으로 이어 붙이듯 연결하기
# 합집합 : 행 인덱스 기준으로 줄을 맞추고, 없는 곳은 NaN
# 열 이름이 같아도(a, b, c) 합치지 않고 둘 다 남음
result3 = pd.concat([df1, df2], axis=1)  # join='outer'
print(result3, '\n')
```

```python
# join='inner' 옵션 적용하기(교집합) → 양쪽에 다 있는 인덱스 2, 3 만 남음
result3_in = pd.concat([df1, df2], axis=1, join='inner')
print(result3_in, '\n')
```

```python
# 시리즈 만들기 (name 이 나중에 열 이름이 됨)
sr1 = pd.Series(['e0', 'e1', 'e2', 'e3'], name='e')                 # 인덱스 0~3
sr2 = pd.Series(['f0', 'f1', 'f2'], name='f', index=[3, 4, 5])      # 인덱스 3~5
sr3 = pd.Series(['g0', 'g1', 'g2', 'g3'], name='g')                 # 인덱스 0~3

print(sr1)
print(sr2)
print(sr3)
```

```python
# df1과 sr1을 좌우 열 방향으로 연결 → e 열이 추가됨 (인덱스 0~3 으로 같아서 NaN 없음)
result4 = pd.concat([df1, sr1], axis=1)
print(result4, '\n')
```

```python
# df2과 sr1을 좌우 열 방향으로 연결
# 인덱스 0~5 합집합, 서로 없는 곳은 NaN / sort=True : 인덱스 정렬
result5 = pd.concat([df2, sr1], axis=1, sort=True)
print(result5, '\n')
```

```python
# 시리즈 2개를 옆으로 → 데이터프레임 (열 e, g)
result6 = pd.concat([sr1, sr3], axis=1)
print(result6, '\n')
```

```python
# 시리즈 2개를 위아래로 → 긴 시리즈 1개 (인덱스 0~3 이 두 번 나옴)
result7 = pd.concat([sr1, sr3], axis=0)
print(result7, '\n')
```

```python
import pandas as pd

# 기업의 기본정보 + 주가
# value : 시가총액
# 주의) 내 폴더에는 ./data/ 에 파일이 있음
df1 = pd.read_excel('./data/stock price.xlsx', engine= 'openpyxl')   # 주가데이터
df1.head()
```

```python
# 투자 판단을 위한 재무 재표
# eps : 주당 순이익(기업당 주당 벌어들이는 이익)
# bps : 주당 순자산가치
# per : 주가수익비율(주가 대비 수익 수준) = 주가 / eps → 낮을수록 저평가
# pbr : 주가자산비율(주가 대비 자산 가치) = 주가 / bps → 1 미만이면 자산보다 싸게 거래
df2 = pd.read_excel('./data/stock valuation.xlsx', engine= 'openpyxl')  # 밸류에이션
df2.head()
```

```python
# 데이터프레임 합치기 - 왼쪽 데이터프레임 기준, 키 값 분리
# key 열 이름이 서로 다를 때 left_on / right_on 으로 각각 지정
# df1 은 전부 남김, 양쪽에 있는 id 열은 id_x(df1), id_y(df2) 로 이름이 바뀜
merge_left = pd.merge(df1, df2, how='left', left_on='stock_name', right_on='name')
print(merge_left)
```

```python
# 데이터프레임 합치기 - 오른쪽 데이터프레임 기준, 키 값 분리
# df2 는 전부 남기고, df1 은 짝이 있는 것만
merge_right = pd.merge(df1, df2, how='right', left_on='stock_name', right_on='name')
print(merge_right)
```

## 5.그룹 연산

#### 5-1.그룹 객체 만들기(분할 단계)

* 1개 열을 기준으로 그룹화

```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','sex', 'class', 'fare', 'survived']]

print('승객 수:', len(df))     # 891
```

```python
# class 기준으로 분할 → 아직 계산 전, 나눠만 놓은 그룹 객체
# 리스트로 넣으면 key 가 튜플 ('First',) 로 나옴
grouped = df.groupby(['class'])
for key, group in grouped:        # 그룹 객체는 for 로 하나씩 꺼낼 수 있음
    print(type(key), type(group))
    print('* key :', key)
    print('* number :', len(group))    # First 216 / Second 184 / Third 491
    print(group.head())
    print('\n')
```

```python
# 연산 메소드 적용
# numeric_only=True : 숫자 열만 평균 (sex 같은 글자 열이 있으면 TypeError)
# survived 는 0/1 이라 평균 = 생존율
average = grouped.mean(numeric_only=True)
print(average)
```

```python
# 개별 그룹 선택 (리스트로 groupby 했으니 key 도 튜플로)
group3 = grouped.get_group(('Third',))
print(group3.head())
group3
```

* 여러 열을 기준으로 그룹화

```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','sex', 'class', 'fare', 'survived']]

print('승객 수:', len(df))
```

```python
# class 열, sex 열을 기준으로 분할 → 3 x 2 = 6개 그룹
# key 는 ('First', 'female') 처럼 튜플
grouped_two = df.groupby(['class', 'sex'])
for key, group in grouped_two:
    print('* key :', key)
    print('* number :', len(group))
    print(group.head())
    print('\n')
```

```python
# 3등석 여성 그룹만 꺼내기
group3f = grouped_two.get_group(('Third','female'))
print(group3f.head())
```

```python
# 그룹 6개 각각의 평균 → 행 인덱스가 (class, sex) 2단 = 멀티 인덱스
average_two = grouped_two.mean(numeric_only=True)
print(average_two)
print('\n')
print(type(average_two))
```

```python
def min_max(x):    # 최대값 - 최소값
    return x.max() - x.min()

# 열마다 적용할 함수를 딕셔너리로 지정 (내가 만든 함수도 가능)
agg_minmax = grouped.agg({
    'age': min_max,
    'fare': min_max,
    'survived': min_max
})

agg_minmax
```

```python
# 여러 함수를 각 열에 동일하게 적용하여 집계
# sex(글자)도 min/max 가 나오는데 알파벳 순이라 의미 없음
agg_all = grouped.agg(['min', 'max'])
print(agg_all.head())
```

```python
# 각 열마다 다른 함수를 적용하여 집계
agg_sep = grouped.agg({'fare':['min', 'max'], 'age':'mean'})
print(agg_sep.head())
```

```python
# 열 하나만 골라서 계산 (grouped['age'].mean() 과 같음)
age_mean = grouped.age.mean(numeric_only=True)
print(age_mean)
```

*  그룹 객체 필터링

```python
# 데이터 개수가 200개 이상인 그룹만을 필터링하여 데이터프레임으로 반환
# x = 그룹 하나(데이터프레임) → First(216), Third(491) 통과 / Second(184) 탈락
# 결과는 요약표가 아니라 통과한 그룹의 원래 행들
grouped_filter = grouped.filter(lambda x: len(x) >= 200)
print(grouped_filter.head())
print('\n')
print(type(grouped_filter))
```

```python
# age 열의 평균이 30보다 작은 그룹만을 필터링하여 데이터프레임으로 반환
# Second(29.9), Third(25.1) 통과 / First(38.2) 탈락
age_filter = grouped.filter(lambda x: x.age.mean() < 30)
print(age_filter.tail())
print('\n')
print(type(age_filter))
```

## 멀티 인덱스

```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','sex', 'class', 'fare', 'survived']]

grouped = df.groupby(['class', 'sex'])    # 2개 열로 그룹 → 결과 인덱스가 멀티 인덱스
```

```python
# class, sex 가 인덱스로 빠져서 남은 열이 전부 숫자 → numeric_only 없어도 됨
gdf = grouped.mean()
print(gdf)
print('\n')
print(type(gdf))
```

```python
# class 값이 First인 행을 선택하여 출력 (바깥 레벨)
print(gdf.loc['First'])
```

```python
# class 값이 First이고, sex 값이 female인 행을 선택하여 출력 → 한 줄(시리즈)
print(gdf.loc[('First', 'female')])
```

```python
# sex 값이 male인 행을 선택하여 출력
# loc 은 바깥 레벨부터만 지정 가능 → 안쪽 레벨(sex)로 고를 땐 xs 사용
print(gdf.xs('male', level='sex'))
```

## 피벗
* 구성 4가지 요소 : 행 인덱스, 열 인덱스, 데이터 값, 데이터 집계 함수

```python
# 등급별, 성별 평균 나이 (groupby 결과를 가로세로 표로 펼친 모양)
pdf1 = pd.pivot_table(df,             # 피벗할 데이터프레임
                      index='class',    # 행 위치에 들어갈 열
                      columns='sex',    # 열 위치에 들어갈 열
                      values='age',     # 데이터로 사용할 열
                      aggfunc='mean')   # 데이터 집계 함수

print(pdf1.head())
```

```python
# class별, 성별, 생존율, 생존자 수 집계(mean, sum), value : survived
# survived 가 0/1 이라 mean = 생존율, sum = 생존자 수
# 함수가 2개라 열 인덱스가 (함수, sex) 2단으로 생김
pdf2 = pd.pivot_table(df,                      # 피벗할 데이터프레임
                      index='class',             # 행 위치에 들어갈 열
                      columns='sex',             # 열 위치에 들어갈 열
                      values='survived',         # 데이터로 사용할 열
                      aggfunc=['mean', 'sum'])   # 데이터 집계 함수

pdf2.head()
```

```python
# class, sex 별 평균나이, 최대 요금
# 열 인덱스 3단 : (함수, 값, survived) → 2 x 2 x 2 = 8개 열

pdf3 = pd.pivot_table(df,                      # 피벗할 데이터프레임
                    index=['class', 'sex'],    # 행 위치에 들어갈 열
                    columns='survived',        # 열 위치에 들어갈 열
                    values=['age', 'fare'],    # 데이터로 사용할 열
                    aggfunc=['mean', 'max'])   # 데이터 집계 함수

pdf3.head()

# 1등급 여성중 생존자의 평균 나이 : 34.939
```

```python
# 행, 열 구조 살펴보기
print(pdf3.index)       # MultiIndex (class, sex)
print(pdf3.columns)     # MultiIndex (함수, 값, survived)
```

```python
# xs 인덱서 사용 - 행 선택(default: axis=0)
# 바깥 레벨 class 가 First 인 행들
print(pdf3.xs('First')) 
```

```python
# 행 인덱스가 ('First', 'female')인 행을 선택
print(pdf3.xs(('First', 'female')))  
```

```python
# 행 인덱스의 sex 레벨이 male인 행을 선택
print(pdf3.xs('male', level='sex')) 
```

```python
 # Second, male인 행을 선택
 # level 은 번호(0)와 이름('sex')을 섞어 써도 됨
print(pdf3.xs(('Second', 'male'), level=[0, 'sex']))
```

```python
# xs 인덱서 사용 - 열 선택(axis=1 설정)
# axis=1 을 안 쓰면 행에서 찾음
print(pdf3.xs('mean', axis=1))
```

```python
# 열 인덱스가 ('mean', 'age')인 데이터 선택 → 사망(0)/생존(1)별 평균 나이
print(pdf3.xs(('mean', 'age'), axis=1)) 
```

```python
# survived 레벨이 1인 데이터 선택 → 생존자 열만
print(pdf3.xs(1, level='survived', axis=1)) 
```

```python
# max, fare, survived=0인 데이터 선택 → 사망자의 최대 요금
# level 을 지정하면 꺼낸 레벨 이름이 결과에 그대로 남음
print(pdf3.xs(('max', 'fare', 0),
              level=[0, 1, 2], axis=1)) 
```
