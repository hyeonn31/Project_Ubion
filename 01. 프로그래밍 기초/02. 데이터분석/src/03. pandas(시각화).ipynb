```python
import pandas as pd
import matplotlib.pyplot as plt

# 시도별 전출입 인구수.xlsx

df = pd.read_excel('./data/시도별 전출입 인구수.xlsx', engine= 'openpyxl', header= 0)
df.head()
```

```python
# 누락값(NaN)을 앞 데이터로 채움
df = df.ffill()
df.head()
```

```python
df['전출지별'].head()
```

```python
# 서울에서 다른 지역으로 이동한 데이터만 추출하여 정리
mask = (df['전출지별'] == '서울특별시')  & (df['전입지별'] != '서울특별시') 
# mask.head()

df_seoul = df[mask]
# df_seoul.head()

# 전출지별 삭제
df_seoul = df_seoul.drop(['전출지별'], axis=1)
# df_seoul.head()

# 전입지별 이름을 전입지로 바꾸기
df_seoul.rename({'전입지별':'전입지'}, axis=1, inplace=True)
df_seoul.set_index('전입지', inplace=True)
df_seoul.head()
```

```python
# 서울에서 경기도로 이동한 인구 데이터 값만 선택 
sr_one = df_seoul.loc['경기도']
sr_one.head()
```

```python
plt.style.use('ggplot')

fig = plt.figure(figsize=(20,5))
ax = fig.add_subplot(1,1,1)

ax.plot(col_years, df_3.loc['충청남도',:],marker='o', markerfacecolor = 'green', \
        markersize =10, color ='olive', linewidth =2, label = '서울 -> 충남')
ax.plot(col_years, df_3.loc['경상북도',:],marker='o', markerfacecolor = 'blue', \
        markersize =10, color ='skyblue', linewidth =2, label = '서울 -> 경북')
ax.plot(col_years, df_3.loc['강원도',:],marker='o', markerfacecolor = 'red', \
        markersize =10, color ='magenta', linewidth =2, label = '서울 -> 강원')

```

```python
# 범례 표시
ax.legend(loc='best')

# 차트 제목 추가
ax.set_title('서울 -> 충남, 경북, 강원 인구 이동', size=20)

# 축이름 추가
ax.set_xlabel('기간', size=12)
ax.set_ylabel('이동 인구수', size = 12)

# 축 눈금 라벨 지정 및 90도 회전
ax.set_xticklabels(col_years, rotation=70)

# 축 눈금 라벨 크기
ax.tick_params(axis="x", labelsize=10)
ax.tick_params(axis="y", labelsize=10)

plt.show()
```

## 경기도 행정구역별 인구 시각화

```python
import pandas as pd
import folium
import json

df= pd.read_excel('./data/경기도인구데이터.xlsx', index_col='구분', engine='openpyxl')
df.columns = df.columns.map(str)
df.head()
```

```python
# 경기도 시군구 경계 정보를 가진 geo-json 파일 불러오기
try:
    geo_data = json.load(open('./data/경기도행정구역경계.json', encoding='utf-8'))
except:
     geo_data = json.load(open('./data/경기도행정구역경계.json', encoding='utf-8-sig'))     # 혹시 BOmM이 있으면 제거 하고 BOM(Byte Order Mark) 제거하고 읽기
```

```python
# 경기도 지도 만들기
g_map = folium.Map(location=[37.5502, 126.982], zoom_start=9)

# 출력할 연도 선택 (2007 ~ 2017년 중에서 선택)
year = '2007'

# Choropleth 클래스로 단계구분도 표시하기
# 행정구역과 같이 지도 상의 어떤 경계에 둘러싸인 영역에 색을 칠하거나, 음영으로 정보를 나타냄

folium.Choropleth(geo_data=geo_data,                # 지도 경계
                  data= df[year],                   # 표시하려는 데이터
                  columns= [df.index, df[year]],    # 열지정
                  fill_color='YlOrRd', fill_opacity=0.7, line_opacity=0.3,
                  threshold_scale=[10000, 100000, 300000, 500000, 700000],
                  key_on='feature.properties.name',
                  ).add_to(g_map)

'''
1만 ~ 10만 : 연한 노랑, 10만~30만 : 노랑 ~ 주황, 30만 ~ 50만 : 주황, 50만 ~ 70만 : 주황 ~ 빨강, 70만 이상 : 진한 빨강
'''

g_map.save('./output/gyonggi_population_' + year + '.html')
```

## 시도별 금융 지표 단계구분도

```python
import requests, folium
import pandas as pd
import numpy as np

geo_url = ("https://raw.githubusercontent.com/southkorea/southkorea-maps/master/kostat/2018/json/skorea-provinces-2018-geo.json")
res = requests.get(geo_url, timeout=10) # requests는 크롤링 할수 있는 패키지 timeout은 10초안에 가져온다.
print(res.status_code)  # 200으로 떨어지면 정상이다.

res.raise_for_status()  # error가 나면 거기서 멈춘다.
geo = res.json()        # json값을 dictionary로 가져온다.


# 샘플 데이터 (임의 값). 2018 파일의 properties.name은 '서울특별시' 같은 정식 명칭
# 가상의 지역별 가계대출 데이터 생성
names = ['서울특별시','부산광역시','대구광역시','인천광역시','광주광역시','대전광역시',
         '울산광역시','세종특별자치시','경기도','강원도','충청북도','충청남도',
         '전라북도','전라남도','경상북도','경상남도','제주특별자치도']

rng = np.random.default_rng(1)  # 시드가 1인 난수를 생성
df= pd.DataFrame({'region' : names, 'loan' : rng.integers(10,300,len(names))}) # loan에 대한 값이 10부터 300사이의 정수 난수를 len(names)만큼 생성

# 지역명 일치 여부 점검 (빈 set이면 정상)
# 먼저 geo에서 feature 키를 가지는 값을 하나씩 꺼내와서 f에 담은뒤 f에서 properties라는 속성에서 name함수를 가져온다.
geo_names = {f['properties']['name'] for f in geo['features']}
# ^ : 양쪽에 모두 있는 지역명은 제외, 한쪽에만 있는 지역명 출력
# 아무것도 안나온다는것은 지역명이 양쪽 그림그릴 데이터와 난수로 만든 대출 데이터에 모두 일치하게 있다는 뜻이다.
print('불일치:', set(df['region']) ^ geo_names)

m = folium.Map(location=[36.3, 127.8], zoom_start=7, tiles='OpenStreetMap')
folium.Choropleth(
    geo_data=geo,
    data =df,
    columns=['region', 'loan'],
    key_on='feature.properties.name',
    fill_color='YlOrRd' ,
    fill_opacity=0.8,
    line_opacity=0.4,
    legend_name='시도별 가계대출 (샘플, 조 원)',
).add_to(m)
```

## 국가간 자금 흐름선 그리기

```python
# 서울에서 뉴욕, 도쿄, 싱가포르로 이동하는 자금 흐름을 세계 지도 위에 선
import folium
m = folium.Map(
    location=[30, 100],
    zoom_start=3,
    tiles='OpenStreetMap'
)

seoul = [37.57, 126.98]
# 목적지, 거래 유혀으 금액(억원)
flows = [
    ([40.71, -74.00], '서울->뉴욕 투자', 500),
    ([35.68, 139.69], '서울->도쿄 송금', 200),
    ([1.35, 103.82], '서울->싱가포르 투자', 800)
]

for dest, label, amount in flows :          # 이제 설명안해도 알겠다. flows에 값이 3개가 하나씩 묶여있다. 그러면 첫번째 좌표가 dest로 제목이 label로 돈이 amount로 할당되어 for문에서 구동
    folium.PolyLine(
        [seoul, dest],
        weight=amount / 100,
        color = 'crimson',
        opacity = 0.7 ,
        tooltip=f'{label}: {amount}억원'
    ).add_to(m)

m
```
