```python
import pandas as pd

df= pd.read_csv('./Data-Pandas/age.csv',encoding='cp949')

df
```

```python
# 남양주에 사는 사람들의 나이별 인구수를 리스트에 저장
result = []   # 나이별 인구수를 담을 빈 리스트

for i in range(len(df)):                    # i = 0, 1, 2, ... (줄 번호)
    name = df.loc[i, '행정구역']            # i 번째 줄의 행정구역 이름
    if '남양주시' in name:                # 남양주시 전체 줄만 고르기 (읍·면·동 줄은 제외)
        for value in df.iloc[i, 3:]:        # 3번 열(0세)부터 끝(100세 이상)까지 하나씩
            value = value.replace(',', '')  # '3,740' → '3740' : 쉼표 떼기
            result.append(int(value))       # 숫자로 바꿔서 리스트에 추가

print(len(result))   # 101 → 0세 ~ 100세 이상
print(result)
```
