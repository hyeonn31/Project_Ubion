## 1. 함수 매핑

#### 1-1. 개별 원소에 함수 매핑

* 시리즈 원소에 함수 매핑

* 데이터프레임 원소에 함수 매핑

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]     # 모든 행, age와 fare 열만 선택
df['ten'] = 10                          # 모든 행에 10이 들어간 새 열 ten 추가
print(df.head())
```

```python
# 사용자 함수 정의
def add_10(n):              # 숫자 n을 받아서 10을 더한 값을 반환
    return n + 10

def add_two_obj(a,b):       # 두 값 a, b를 받아서 더한 값을 반환
    return a+b
```

```python
# 함수가 잘 동작하는지 숫자로 먼저 테스트
print(add_10(10))           # 10 + 10 = 20
print(add_two_obj(10,10))   # 10 + 10 = 20
```

```python
#시리즈 객체에 적용
# apply(함수) : age 열의 원소를 하나씩 꺼내 add_10에 넣고, 반환값을 모아 새 시리즈로 만듦
# 함수 이름만 넣음 (add_10() 처럼 괄호를 붙이면 안 됨)
sr1 = df['age'].apply(add_10)
print(sr1.head())
```

```python
# 시리즈 객체와 숫자에 적용 : 2개의 인수 (시리즈 + 숫자)
# 첫 번째 인수 a 에는 age 열의 원소가 하나씩 자동으로 들어가고,
# 나머지 인수는 b=10 처럼 키워드로 따로 전달
sr2 = df['age'].apply(add_two_obj, b=10)
print(sr2.head())
```

```python
# 람다 함수 활용 : 시리즈 객체에 적용
# lambda x: add_10(x) : 이름 없는 한 줄짜리 함수 (x를 받아 add_10(x)를 반환)

sr3 = df['age'].apply(lambda x: add_10(x)) # x : age 열의 원소 하나 (22.0, 38.0, ...)
# sr3 = df['age'].apply(lambda x: add_10(x) if x % 2 == 0 else x)   # 짝수일 때만 10을 더하고, 아니면 그대로
print(sr3.head())
```

```python
import pandas as pd

df = pd.DataFrame({
    '종목': ['삼성전자', 'SK하이닉스', '현대차', '카카오'],
    '종가': [70000, 180000, 250000, 55000],
    '수익률(%)': [2.5, -1.2, 3.8, -0.5],
    '거래량': [1000000, 800000, 450000, 1200000]
})

print(df)
```

** 시리즈 원소에 함수 매핑

```python
# 예제 1. 종가에 10% 목표 수익률 적용
def target_price(price):        # 가격을 받아서 10% 올린 가격을 반환
    return price * 1.10

# 종가 시리즈에 함수 매핑: '목표가' 변수명 추가
# 종가 열의 원소를 하나씩 target_price에 넣고, 결과를 새 열 '목표가'로 추가
df['목표가'] = df['종가'].apply(target_price)
# df['목표가'] = df['종가'].map(target_price)

df.head()
```

```python
# 예제 2. 수익률에 따라 매수 . 매도 신호 만들기
def trading_signal(rate):
    if rate >=3:
        return '강한상승'
    elif rate >0:
        return '상승'
    elif rate <=-1:
        return '하락'
    else:
        return '약보합'
    
    
df['시장신호'] = df['수익률(%)'].apply(trading_signal)
df.head()
    
```

```python
# 예제 3. 종가가 100,000원 이상이면 목표가 계산, 미만이면 0
# 목표가 계산 함수 : 종가 대비 10% 상승

def target_price(price):
        return price* 1.10
   

df['목표가']=df['종가'].apply(
    lambda x : target_price(x) if x >= 100000 else 0
)

df.head()
```

```python
# 예제 5. 데이터프레임의 숫자를 등급으로 변환

scores = pd.DataFrame({
    '삼성전자': [90, 65, 30],
    'SK하이닉스': [75, 95, 45]
}, index=['상승점수', '안정점수', '위험점수'])

def grade(score):
    if score >= 80:
        return 'A'
    elif score >= 60:
        return 'B'
    else :
        return 'C'


result = scores.map(grade)
result
```

#### 1-2. 시리즈 객체에 함수 매핑

* 데이터프레임의 각 열에 함수 매핑

** 리턴값이 시리즈

```python
import seaborn as sns

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]     # 모든 행, age와 fare 열만 선택                        # 모든 행에 10이 들어간 새 열 ten 추가
print(df.head())
```

```python
# 사용자 함수 정의
def missing_value(series):  # 시리즈를 인수로 전달
    return series.isnull()  # 불린 시리즈를 반환

# 데이터프레임의 각 열을 인수로 전달하면 데이터프레임을 반환
result = df.apply(missing_value, axis=0)
print(result.head())
print('\n')
print(type(result))
```

** 리턴값이 원소

```python
def min_max(x):
    return x.max() - x.min()

#데이터프레임의 각 열을 (시리즈) 인수로 전달하면 시리즈를 반환
result = df.apply(min_max)
print(result)
print('\n')
print(type(result))
```

####  1-3. 데이터프레임 객체에 함수 매핑

```python
import seaborn as sns

titanic = sns. load_dataset('titanic')
df = titanic. loc[:, ['age','fare']]
df['ten'] = 10
print(df.head())
```

```python
# 각 열의 NaN 찾기 - 데이터프레임 전달하면 데이터프레임을 반환
def missing_value(x):
    return x.isnull()

# 각 열의 NaN 개수 반환 - 데이터프레임 전달하면 시리즈 반환
def missing_count(x):
    return missing_value(x).sum()

# 데이터프레임의 총 NaN 개수 - 데이터프레임 전달하면 값을 반환
def total_number_missing(x):
    return missing_count(x).sum()

result_df = df.pipe(missing_value)
print(result_df.head())
print(type(result_df))
```

```python
result_series = df.pipe(missing_count)
print(result_series)
print(type(result_series))
```

```python
result_value = df.pipe(total_number_missing)
print(result_value)
print(type(result_value))
```
