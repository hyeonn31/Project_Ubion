* 네이버 금융 데이터 수집

```python
import time 
import os
import pandas as pd
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.service import Service

service = Service()  # Selenium 서비스 생성
options = webdriver.ChromeOptions()  # Chrome 브라우저 옵션 설정
driver = webdriver.Chrome(service=service, options=options) # Chrome 브라우저 실행 객체

#  네이버 증권 페이지 접속
url = "https://stock.naver.com/domestic/stock/005930/price"
driver.get(url)

wait = WebDriverWait(driver, 15)
rows = wait.until(
    EC.presence_of_all_elements_located(
        (By.CSS_SELECTOR, "table tbody tr")
    )
)
```

```python
result = []

for row in rows:
    cells = row.find_elements(By.CSS_SELECTOR, "td")
    # for cell in cells:
    #     print(cell.text.strip())

    # 데이터가 없는 행 제외
    if len(cells) < 6:
        continue

    values = [
        cell.text.strip()
        for cell in cells
    ]

    # print(values)
    result.append({
        "날짜": values[0],
        "종가": values[1],
        "전일비": values[2],
        "시가": values[3],
        "고가": values[4],
        "저가": values[5]
    })

# DataFrame 생성
df = pd.DataFrame(result)

# CSV 저장
df.to_csv(
    "samsung_stock.csv",
    index=False,
    encoding="utf-8-sig"
)

print(df)
print("CSV 저장 완료: samsung_stock.csv")
```

## Project(쿠팡 데이터 수집)
* https://www.coupang.com/np/categories/194276 : 데이터 수집(Mission)

```python
import undetected_chromedriver as uc
from selenium.webdriver.chrome.service import Service # ChromeDriver 실행을 관리하는 객체 
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys 
import time
import pyperclip
import requests

options = uc.ChromeOptions()
options.add_argument("--start-maximized")

driver = uc.Chrome(
    options=options,
    version_main=155
)

url = "https://www.coupang.com/np/categories/194276"
driver.get(url)
```

```python
productList = driver.find_element(By.ID, 'product-list')
lis= productList.find_elements(By.TAG_NAME, 'li')
print(len(lis))
```

```python
prd_name = []
prd_price = []
prd_deli = []
prd_url = []
prd_foodname = []


for idx, li in enumerate(lis):
    try:
        name = li.find_element(By.CLASS_NAME, 'ProductUnit_productNameV2__cV9cw').text
        price = li.find_element(By.CLASS_NAME, 'Price_priceValue__A4KOr').text
        deli = li.find_element(By.CLASS_NAME, 'DeliveryInfo_delivery__c7z4P').text
        img_url = li.find_element(By.TAG_NAME, 'img').get_attribute('src')
        url = li.find_element(By.TAG_NAME, 'a').get_attribute('href')
        
        # 이미지 다운로드
        img_res = requests.get(img_url)
        img_res.raise_for_status()

        food_name = "./image/coupang/food{}.jpg".format(idx + 1)
        with open(food_name, "wb") as f:
            f.write(img_res.content)

        prd_name.append(name)
        prd_price.append(price)
        prd_deli.append(deli)
        prd_url.append(url)
        prd_foodname.append(food_name)
        
    except Exception as e:
        print(f"{idx+1}번째 상품 건너뜀: {e}")
        continue

final = pd.DataFrame({'이름': prd_name, '가격': prd_price, 'Delivery': prd_deli, \
                      'URL': prd_url, '리소스명' : prd_foodname})
final.to_csv('쿠팡식품.csv', encoding='cp949')  # 'utf-8' euc-kr 
final
    
```
