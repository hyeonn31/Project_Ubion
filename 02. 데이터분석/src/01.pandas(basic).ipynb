## 시리즈

* dict to series

```python
import pandas as pd   # pandas 를 pd 라는 짧은 이름으로 불러오기

# 딕셔너리 → 시리즈 : 키(key)는 인덱스, 값(value)은 데이터가 된다
dict_data = {'a': 1, 'b': 2, 'c': 3}

sr = pd.Series(dict_data)   # 딕셔너리로 시리즈 만들기
print(type(sr))             # <class 'pandas.core.series.Series'>
print()                     # 빈 줄 출력
print(sr)                   # 왼쪽 = 인덱스(a, b, c), 오른쪽 = 값(1, 2, 3)
```

* list to series - series index

```python
import pandas as pd

# 리스트 → 시리즈 : 인덱스를 따로 안 주면 0, 1, 2 ... 정수 인덱스가 자동으로 붙는다
list_data = ['2019-01-02', 3.14, 'ABC', 100, True]   # 자료형이 섞여 있어도 OK (dtype = object)
sr = pd.Series(list_data)

print(sr)
```

```python
idx = sr.index     # 인덱스(이름표)만 꺼내기 → RangeIndex(start=0, stop=5, step=1)
val = sr.values    # 값만 꺼내기 → numpy 배열
print(idx)
print('\n')        # '\n' = 줄바꿈 (빈 줄을 넣어 결과를 구분)
print(val)
```

* tuple to series

```python
import pandas as pd

# 튜플 → 시리즈 : index= 로 인덱스 이름을 직접 지정할 수 있다 (값 개수와 같아야 함)
tup_data = ('영인', '2010-05-01', '여', True)
sr = pd.Series(tup_data, index=['이름', '생년월일', '성별', '학생여부'])
print(sr)
```

```python
# 원소를 1개 선택
print(sr.iloc[0])     # iloc = 위치(숫자) 로 선택 → 0번째 값 '영인'
print(sr['이름'])     # 인덱스 이름으로 선택 → '영인'
print('\n')
```

```python
# 여러 개의 원소를 선택
print(sr.iloc[1 : 3])          # 1번 ~ 2번 위치 (iloc 슬라이싱은 끝 번호 3 미포함)
print('\n')
print(sr[['생년월일', '성별']])  # 이름 여러 개는 리스트 [ ] 로 한 번 더 감싼다
```

## 데이터프레임

* dict to dataframe

```python
import pandas as pd

# 딕셔너리 → 데이터프레임 : 키는 열 이름, 값(리스트)은 그 열의 데이터가 된다
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}

df = pd.DataFrame(dict_data)   # 3행 x 5열 표. 행 인덱스는 0, 1, 2 자동 생성
print(type(df))                # <class 'pandas.core.frame.DataFrame'>
print('\n')
print(df)
```

* list to dataframe

```python
import pandas as pd

# 2차원 리스트 → 데이터프레임 : 안쪽 리스트 하나가 '행' 하나
df = pd.DataFrame([[15, '남', '덕영중'], [17, '여', '수리중']],
                   index=['준서', '예은'],              # 행 이름
                   columns=['나이', '성별', '학교'])    # 열 이름

print(df)            #데이터프레임
print('\n')
print(df.index)      #행 인덱스
print('\n')
print(df.columns)    #열 이름
print('\n')
```

* 열, 행 이름 변경

```python
import pandas as pd

df = pd.DataFrame([[15, '남', '덕영중'], [17, '여', '수리중']],
                   index=['준서', '예은'],
                   columns=['나이', '성별', '학교'])

print(df)            #데이터프레임
print('\n')

# rename(columns={기존이름: 새이름}) → 열 이름 변경
#   inplace=True : 원본 df 를 직접 바꿈 (반환값 없음)
df.rename(columns={'나이':'연령', '성별':'남녀', '학교':'소속'}, inplace=True)

# rename(index={...}) → 행 이름 변경
#   inplace 없이 쓰면 새 데이터프레임을 돌려주므로 df = ... 로 다시 저장해야 한다
df = df.rename(index={'준서':'학생1', '예은':'학생2' })

print(df)
```

```python
import pandas as pd

# 행 선택 실습용 데이터 (위 셀에서 행을 지웠을 수 있으니 새로 만든다)
exam_data = {'수학' : [ 90, 80, 70], '영어' : [ 98, 89, 95],
             '음악' : [ 85, 95, 100], '체육' : [ 100, 90, 90]}

df = pd.DataFrame(exam_data, index=['서준', '우현', '인아'])
print(df)
print('\n')
```

```python
# 행 인덱스를 사용하여 행 1개를 선택 : loc(문자 인덱스), iloc(숫자 인덱스)
# 서준 행 검색 : loc 인덱서 활용
label1 = df.loc['서준']      # 결과는 시리즈 (열 이름이 인덱스가 됨)

# 0번째 행 검색 : iloc 인덱서 활용
position1 = df.iloc[0]       # 0번째 행 = 서준 → label1 과 같은 결과

print(label1)
print('\n')
print(position1)
```

```python
# 행 인덱스를 사용하여 2개 이상의 행 선택
# 서준, 우현 : 인덱싱(loc, iloc 사용)  → 이름/번호를 리스트 [ ] 로 감싼다
label2 = df.loc[['서준', '우현']]
position2 = df.iloc[[0, 1]]
print(label2)
print('\n')
print(position2)
print('\n')

# 서준, 우현 : 슬라이싱(loc, iloc 사용)
label3 = df.loc['서준':'우현']   # loc 슬라이싱은 끝('우현') 포함
position3 = df.iloc[0:2]         # iloc 슬라이싱은 끝(2) 미포함 → 0, 1 행
print(label3)
print('\n')
print(position3)
```

* Transpose

```python
import pandas as pd

# 전치(행 ↔ 열 바꾸기) 실습용 데이터
exam_data = {'이름' : [ '서준', '우현', '인아'],
             '수학' : [ 90, 80, 70],
             '영어' : [ 98, 89, 95],
             '음악' : [ 85, 95, 100],
             '체육' : [ 100, 90, 90]}
df = pd.DataFrame(exam_data)
print(df)
```

```python
# transpose() : 행과 열을 맞바꾼다 (inplace 옵션이 없으므로 df = ... 로 저장)
df = df.transpose()
df
```

```python
# df.T : transpose() 의 줄임말. 한 번 더 바꾸면 원래 모양으로 돌아온다
df = df.T
df
```

* 멀티 인덱스(금융)

```python
# 날짜별 · 종목별 종가 데이터 (같은 날짜에 종목이 여러 개)
data = pd.DataFrame({
    'Date': ['2026-02-01','2026-02-01','2026-02-02','2026-02-02'],
    '종목': ['삼성','현대','삼성','현대'],
    'Close': [100, 200, 102, 198]
})

data
```

* reindex

```python
import pandas as pd

# reindex 실습용 데이터 : 행 이름 r0, r1, r2
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}

df = pd.DataFrame(dict_data, index=['r0', 'r1', 'r2'])
df
```

```python
# 인덱스를 [r0, r1, r2, r3, r4]로 재지정
# reindex : 원하는 인덱스 목록으로 행을 다시 배치
#   기존에 없던 r3, r4 는 채울 값이 없어서 NaN(빈 값)이 된다
new_index = ['r0', 'r1', 'r2', 'r3', 'r4']
ndf = df.reindex(new_index)
ndf
```

```python

# reindex로 발생한 NaN값을 숫자 0으로 채우기
new_index = ['r0', 'r1', 'r2', 'r3', 'r4']
ndf2 = df.reindex(new_index, fill_value=0)   # fill_value = 새로 생긴 칸에 넣을 값
print(ndf2)
```

```python
# 행 인덱스를 정수형으로 초기화
# reset_index : 기존 인덱스(r0, r1, r2)는 'index' 라는 열로 내려가고, 0, 1, 2 가 새 인덱스
ndf = df.reset_index()
ndf
```

* sort index

```python
import pandas as pd

# 정렬 실습용 데이터 (c3 의 첫 값을 41 로 바꿔 값 정렬 결과가 눈에 띄게)
dict_data = {'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[41,11,12], 'c4':[13,14,15]}

df = pd.DataFrame(dict_data, index=['r0', 'r1', 'r2'])
df
```

```python
# 내림차순으로 행 인덱스 정렬
# sort_index : 행 이름(인덱스) 기준 정렬 / ascending=False → 내림차순 (r2, r1, r0)
ndf = df.sort_index(ascending=False)
print(ndf)
```

```python
ndf = df.sort_index()  # 기본값은 오름차순 (ascending=True 생략)
print(ndf)
```

```python
# c3 열을 기준으로 내림차순 정렬
# sort_values(by='열이름') : 인덱스가 아니라 '값' 기준 정렬 → 41, 12, 11 순서
ndf = df.sort_values(by='c3', ascending=False)
print(ndf)
```

* series to series

```python
import pandas as pd

# 두 학생의 점수 : 과목 순서가 서로 다르다 (연산할 때 순서는 상관없음)
student1 = pd.Series({'국어':100, '영어':80, '수학':90})
student2 = pd.Series({'수학':80, '국어':90, '영어':80})

print(student1)
print('\n')
print(student2)
print('\n')
```

```python
# 두 학생의 과목별 점수로 사칙연산 수행
#   ※ 시리즈끼리 연산하면 '순서'가 아니라 '같은 인덱스(과목 이름)'끼리 계산된다
#     student2 는 수학, 국어, 영어 순서지만 → 국어는 국어끼리, 수학은 수학끼리 계산됨
addition = student1 + student2               #덧셈
subtraction = student1 - student2            #뺄셈
multiplication = student1 * student2         #곱셈
division = student1 / student2               #나눗셈
print(type(division))                        # 결과도 시리즈 → <class 'pandas.Series'>
print('\n')
```

```python
# 사칙연산 결과를 데이터프레임으로 합치기 (시리즈 -> 데이터프레임)
# 시리즈 하나가 행 하나가 되고, 과목 이름이 열 이름이 된다
result = pd.DataFrame([addition, subtraction, multiplication, division],
                      index = ['덧셈','뺄셈','곱셈','나눗셈'])   # 행 이름
result
```

* series to seires2

```python
import pandas as pd
import numpy as np   # np.nan(빈 값)을 쓰기 위해 numpy 불러오기

# 빈 값(NaN)과 짝이 없는 인덱스가 있는 경우
student1 = pd.Series({'국어':np.nan, '영어':80, '수학':90})   # 국어 = NaN
student2 = pd.Series({'수학':80, '국어':90})                  # 영어 없음

print(student1)
print('\n')
print(student2)
print('\n')

# NaN 과 연산하거나, 한쪽에만 있는 인덱스(영어)는 결과가 모두 NaN
addition = student1 + student2               #덧셈
subtraction = student1 - student2            #뺄셈
multiplication = student1 * student2         #곱셈
division = student1 / student2               #나눗셈
print(type(division))                        # 결과도 시리즈 → <class 'pandas.Series'>
print('\n')

result = pd.DataFrame([addition, subtraction, multiplication, division],
                      index = ['덧셈','뺄셈','곱셈','나눗셈'])
result   # 국어, 영어 열은 NaN / 수학만 계산됨
```

```python
import pandas as pd
import numpy as np

student1 = pd.Series({'국어':np.nan, '영어':80, '수학':90})
student2 = pd.Series({'수학':80, '국어':90})

print(student1)
print('\n')
print(student2)
print('\n')

# 연산 메소드 + fill_value=0 : 값이 없는 쪽(NaN 또는 짝 없는 인덱스)을 0 으로 보고 계산
sr_add = student1.add(student2, fill_value=0)   # 덧셈 (+)
sr_sub = student1.sub(student2, fill_value=0)   # 뺄셈 (-)
sr_mul = student1.mul(student2, fill_value=0)   # 곱셈 (*)
sr_div = student1.div(student2, fill_value=0) #인피니티 에러(0으로 나눠서) → 영어 80/0 = inf (에러 아님)

result = pd.DataFrame([sr_add,sr_sub,sr_mul,sr_div],
                      index=['덧셈','뺄셈','곱셈','나눗셈'])

result
```

```python
import pandas as pd
import seaborn as sns   # seaborn : 시각화 라이브러리. 연습용 데이터셋도 제공

# 타이타닉 승객 데이터 불러오기 (891행 x 15열, 인터넷 연결 필요)
titanic = sns.load_dataset('titanic')
titanic
```

```python
# df에 age, fare 추출
df = titanic.loc[:, ['age', 'fare']]   # : = 모든 행, ['age','fare'] = 선택할 열 (나이, 요금)
print(df.head())                              # 앞 5행만 미리보기
print("\n")
print((type(df)))                             # DataFrame
```

```python
#데이터 프레임에 숫자 10 더하기
addition = df + 10   # 모든 칸에 10 이 더해진다 (NaN 은 그대로 NaN)
print(addition.head())
print("\n")
print(type(addition))   # 결과도 DataFrame
```
