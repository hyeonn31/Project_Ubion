### 02-1. 서울시 스타벅스 매장 목록 데이터 수집 : 크롤링
* www.istarbucks.co.kr

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

service = Service()
options = webdriver.ChromeOptions()
driver = webdriver.Chrome(service=service, options=options)

url = 'https://www.istarbucks.co.kr/store/store_map.do?disp=locale'
driver.get(url)
```

```python
# 서울’ 버튼 요소를 찾아 클릭
seoul_btn = "#container > div > form > fieldset > div > section > article.find_store_cont > article > article:nth-child(4) > div.loca_step1 > div.loca_step1_cont > ul > li:nth-child(1) > a"

try:
    WebDriverWait(driver, 20).until(EC.element_to_be_clickable\
                                    ((By.CSS_SELECTOR, seoul_btn))).click()
except TimeoutError:
    print("해당 페이지가 20초 내에 열리지 않았음")
```

```python
# 전체’ 버튼 요소를 찾아 클릭
all_btn = "#mCSB_2_container > ul > li:nth-child(1) > a"
try:
    WebDriverWait(driver, 20).until(EC.element_to_be_clickable \
                                ((By.CSS_SELECTOR, all_btn))).click()
except TimeoutError:
    print("해당 페이지가 20초 내에 열리지 않았음")
```

```python
# BeautifulSoup으로 HTML 파서 만들기
html = driver.page_source  # 크롤링한 HTML 문서 
soup = BeautifulSoup(html, 'html.parser')
html
```

```python
starbucks_soup_list = soup.select('li.quickResultLstCon')
print(len(starbucks_soup_list))
```

```python
#  test code
starbucks_store = starbucks_soup_list[0]
name = starbucks_store.select('strong')[0].text.strip()
lat = starbucks_store['data-lat'].strip()
lng = starbucks_store['data-long'].strip()
store_type = starbucks_store.select('i')[0]['class'][0]
add_tel = str(starbucks_store.select('p.result_details')[0]).split('<br/>')
address = add_tel[0].split('>')[1]
tel = add_tel[1].split('<')[0]

name, lat, lng, store_type, address, tel
```

```python
# test code
for i in range(len(starbucks_soup_list)):
    print(starbucks_soup_list[i]["data-name"])
```

```python
# 서울시 스타벅스 매장 목록 데이터 만들기
starbucks_list = []

for starbucks_store in starbucks_soup_list:
    name = starbucks_store.select('strong')[0].text.strip()
    lat = starbucks_store['data-lat'].strip()
    lng = starbucks_store['data-long'].strip()
    store_type = starbucks_store.select('i')[0]['class'][0]
    add_tel = str(starbucks_store.select('p.result_details')[0]).split('<br/>')
    address = add_tel[0].split('>')[1]
    tel = add_tel[1].split('<')[0]

    starbucks_list.append( [ name, lat, lng, store_type, address, tel])
```

```python
# 데이터프레임 생성
columns = ['매장명','위도','경도','매장타입', '주소','전화번호']
seoul_starbucks_df = pd.DataFrame(starbucks_list, columns = columns)
seoul_starbucks_df.head()
```

```python
# 요약 정보 확인
seoul_starbucks_df.info()
```

```python
seoul_starbucks_df.to_excel('./Data/seoul_starbucks.xlsx', index=False)
```

#### 2-2-1. 서울시 주민등록 인구(구별) 데이터 분석

```python
import pandas as pd

sgg_pop_df = pd.read_csv('./Data/report.txt', sep='\t', header=1)
sgg_pop_df.head()
```

```python
# 서울시 시군구 인구 데이터 살펴보기 
columns = { 
    '시점': 'GIGAN',        #   데이터 기준 연월
    '동별(1)': 'SUM',       # 	상위 행정구역명
    '동별(2)': 'JACHIGU',   # 	자치구명 (예: 종로구, 중구 등)
    '소계': 'SEDAE',        # 	세대 수 (가구 수)
    '소계.1': 'GYE_1',      # 	인구 합계 (전체)
    '남자': 'NAMJA_1',  	#   남자 인구 (전체)
    '여자': 'YEOJA_1',      # 	여자 인구 (전체)
    '소계.2': 'GYE_2',      #   한국인 인구 합계
    '남자.1': 'NAMJA_2',    # 	한국인 남자 인구
    '여자.1': 'YEOJA_2',    #   한국인 여자 인구
    '소계.3': 'GYE_3',      #   등록외국인 인구 합계
    '남자.2': 'NAMJA_3',    #   등록외국인 남자 인구
    '여자.2': 'YEOJA_3',    #   등록외국인 여자 인구
    '소계.4': 'SEDAEDANGINGU',          # 	세대당 인구 수 (GYE_1 / SEDAE)
    '소계.5': 'N_65SEISANGGORYEONGJA'   # 	65세 이상 고령자 수
} 

sgg_pop_df.rename(columns = columns, inplace = True) 
sgg_pop_df.head() 
```

```python
sgg_pop_df.info()
```

```python
# 필요없는 데이터 제거 : '소계'
condition = sgg_pop_df['JACHIGU'] != '소계'
sgg_pop_df_selected = sgg_pop_df[condition]
sgg_pop_df_selected.head()
```

```python
# 분석에 필요한 칼럼 선택 : JACHIGU, GYE_1

columns = ['JACHIGU', 'GYE_1']
sgg_pop_df_final = sgg_pop_df_selected[columns]
sgg_pop_df_final.head()
```

```python
# 데이터프레임의 칼럼명 변경 : ['시군구명', '주민등록인구']

sgg_pop_df_final.columns = ['시군구명', '주민등록인구']
sgg_pop_df_final.info()
sgg_pop_df_final.head()
```

```python
# 엑셀로 저장
sgg_pop_df_final.to_excel('./data/sgg_pop.xlsx', index=False)
```

#### 2-2-2. 서울시 사업체현황(산업대분류별/동별) 데이터 분석

```python
sgg_biz_df = pd.read_csv('./data/report1.txt', sep='\t', header=3)
sgg_biz_df
```

```python
columns = {
    '시점': 'GIGAN',
    '동별(1)': 'GIGAN',
    '동별(2)': 'JACHIGU',
    '동별(3)': 'DONG',
    '소계': 'SAEOPCHESU_1',
    '여성대표자': 'YEOSEONGDAEPYOJA',
    '소계.1': 'GYE',
    '남': 'NAM',
    '여': 'YEO'
}

sgg_biz_df.rename(columns = columns, inplace = True)
sgg_biz_df.head(10)
```

```python
sgg_biz_df[['DONG', 'SAEOPCHESU_1']]
```

```python
# 시군구별 사업체 현황 데이터 추출(행)
condition = sgg_biz_df['DONG'] == '소계'
sgg_biz_df_selected = sgg_biz_df[condition]
sgg_biz_df_selected.head(10)
sgg_biz_df_selected
```

```python
# 필요없는 칼럼 제거(열)
columns = ['JACHIGU', 'GYE', 'SAEOPCHESU_1']
sgg_biz_df_final = sgg_biz_df_selected [columns]
sgg_biz_df_final.columns = ['시군구명', '종사자수','사업체수']
sgg_biz_df_final.head()
```

```python
# 데이터프레임의 인덱스 초기화
sgg_biz_df_final = sgg_biz_df_final.reset_index(drop=True)
sgg_biz_df_final.head()
```

```python
sgg_biz_df_final.to_excel('./data/sgg_biz.xlsx', index=False)
```
